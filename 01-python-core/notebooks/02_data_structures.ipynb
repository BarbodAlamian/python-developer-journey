## 2. What Are Data Structures?

A data structure is a way to organize and store data so that we can access and modify it efficiently.

Python provides four important built-in data structures:

- `list`
- `tuple`
- `set`
- `dict`

```python
numbers = [10, 20, 30]
coordinates = (35.7, 51.4)
unique_numbers = {10, 20, 30}
student = {"name": "Barbod", "age": 23}

print(numbers)
print(coordinates)
print(unique_numbers)
print(student)
```

## 3. Lists

A list is an ordered and mutable collection of items.

- **Ordered:** Items maintain their order.
- **Mutable:** Items can be changed after creation.
- **Duplicates allowed:** The same value can appear multiple times.
- **Indexed:** Items can be accessed using their indexes.

```python
scores = [18, 15, 20, 18]

print(scores)
print(type(scores))
```

## 4. List Indexing

Python uses zero-based indexing, which means the first item has index `0`.

The index of each item determines its position in the list.

```python
scores = [18, 15, 20, 12]

print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
```

```python
scores[3]
```

## 5. Negative Indexing

Negative indexing allows us to access items from the end of a list.

- `-1` refers to the last item.
- `-2` refers to the second-to-last item.

```python
scores = [18, 15, 20, 12]

print(scores[-1])
print(scores[-2])
print(scores[-3])
```

## 6. List Slicing

Slicing allows us to extract a portion of a list.

Syntax:

`list[start:stop:step]`

- `start`: The index where slicing begins.
- `stop`: The index where slicing stops, excluding that index.
- `step`: The interval between selected items.

```python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
print(numbers[::])
print(numbers[::1])
print(numbers[::-1])
```

## 7. Changing List Items

Lists are mutable, so we can change individual items using their indexes.

```python
scores = [18, 15, 20]

scores[1] = 17

print(scores)
```

## 8. Adding Items with `append()`

The `append()` method adds one item to the end of a list.

Syntax:

`list.append(item)`

```python
scores = [18, 15, 20]

scores.append(19)

print(scores)
```

## 9. Adding Items with `insert()`

The `insert()` method adds an item at a specific index.

Syntax:

`list.insert(index, item)`

```python
numbers = [10, 20, 40]

numbers.insert(2, 30)

print(numbers)
```

## 10. Removing Items

Python provides several ways to remove items from a list:

- `remove(value)` removes the first matching value.
- `pop(index)` removes and returns an item.
- `del` removes an item or a slice by index.

```python
numbers = [10, 20, 30, 20, 40]

numbers.remove(20)
print(numbers)
```

```python
numbers = [10, 20, 30, 40]

removed_item = numbers.pop(1)

print(removed_item)
print(numbers)
```

```python
numbers = [10, 20, 30]

last_item = numbers.pop()

print(last_item)
print(numbers)
```

```python
numbers = [10, 20, 30, 40]

del numbers[1]

print(numbers)
```

## 11. List Length with `len()`

The `len()` function returns the number of items in a list.

Syntax:

`len(list)`

```python
scores = [18, 15, 20, 12, 17]

number_of_scores = len(scores)

print(number_of_scores)
```

## 12. Checking Items with `in`

The `in` operator checks whether an item exists in a list.

It returns a Boolean value:

- `True` if the item exists.
- `False` if the item does not exist.

```python
scores = [18, 15, 20, 12]

print(20 in scores)
print(10 in scores)
```

```python
scores = [18, 15, 20, 12]

has_perfect_score = 20 in scores

print(has_perfect_score)
```

## 13. Looping Through a List

A `for` loop allows us to process each item in a list.

The loop assigns each item to a variable, one at a time.

```python
scores = [18, 15, 20, 12]

for score in scores:
    print(score)
```

```python
prices = [100, 200, 300]

for price in prices:
    discounted_price = price * 0.9
    print(discounted_price)
```

## 14. Useful List Methods

Python provides several built-in list methods.

- `count(value)` returns how many times a value appears.
- `index(value)` returns the index of the first matching value.
- `extend(iterable)` adds multiple items to a list.
- `clear()` removes all items from a list.

```python
numbers = [10, 20, 30, 20, 40]

print(numbers.count(20))
print(numbers.index(30))
```

```python
numbers = [10, 20, 30]

numbers.extend([40, 50])

print(numbers)
```

```python
numbers = [10, 20, 30]

numbers.append([40, 50])
print(numbers)
```

```python
numbers = [10, 20, 30]

numbers.clear()

print(numbers)
```

## 15. Sorting Lists

Python provides two common ways to sort data:

- `list.sort()` sorts the original list.
- `sorted(list)` returns a new sorted list.

By default, both sort values in ascending order.

```python
numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)
```

```python
numbers = [40, 10, 30, 20]

numbers.sort(reverse=True)

print(numbers)
```

```python
numbers = [40, 10, 30, 20]

sorted_numbers = sorted(numbers)

print(sorted_numbers)
print(numbers)
```

## 16. Reversing a List

The `reverse()` method reverses the order of items in the original list.

It does not sort the values. It simply reverses their current order.

```python
numbers = [30, 10, 40, 20]

numbers.reverse()

print(numbers)
```

## 17. Copying Lists

Assigning a list to another variable does not create a separate list.

Both variables refer to the same list object.

To create a separate shallow copy, we can use:

- `list.copy()`
- `list()`
- List slicing with `[:]`

```python
original = [10, 20, 30]
another = original

another.append(40)

print(original)
print(another)
```

```python
original = [10, 20, 30]
copied = original.copy()

copied.append(40)

print(original)
print(copied)
```

## 19. Introduction to Tuples

A tuple is an ordered and immutable collection of items.

- **Ordered:** Items maintain their order.
- **Immutable:** Items cannot be reassigned after the tuple is created.
- **Duplicates allowed:** The same value can appear multiple times.
- **Indexed:** Items can be accessed using indexes.

Tuples are useful when we want to keep a collection of values unchanged.

```python
coordinates = (35.7, 51.4)

print(coordinates)
print(type(coordinates))
print(coordinates[0])
print(coordinates[1])
```

## 20. Tuple Immutability

Tuples are immutable, so we cannot reassign an individual item after creation.

Attempting to change an item raises a `TypeError`.

```python
coordinates = (35.7, 51.4)

# coordinates[0] = 36.0 --> TypeError: 'tuple' object does not support item assignment
```

## 21. Creating Tuples

Tuples are commonly created using parentheses.

A tuple with one item requires a trailing comma.

```python
point = (10, 20)
single_value_tuple = (10,)

not_a_tuple = (10)

print(type(point))
print(type(single_value_tuple))
print(type(not_a_tuple))
```

## 22. Tuple Unpacking

Tuple unpacking allows us to assign tuple items to separate variables.

The number of variables must match the number of items being unpacked.

```python
coordinates = (35.7, 51.4)

latitude, longitude = coordinates

print(latitude)
print(longitude)
```

```python
x, y = 10, 20

print(x)
print(y)
```

## 23. Returning Multiple Values

A Python function can return multiple values.

When multiple values are returned, Python packs them into a tuple.

We can use tuple unpacking to assign the returned values to separate variables.

```python
def calculate_statistics(numbers):
    minimum = min(numbers)
    maximum = max(numbers)

    return minimum, maximum


scores = [18, 15, 20, 12]

result = calculate_statistics(scores)

print(result)
print(type(result))
```

```python
scores = [18, 15, 20, 12]

minimum, maximum = calculate_statistics(scores)

print("Minimum:", minimum)
print("Maximum:", maximum)
```

## 24. Useful Tuple Methods

Tuples have fewer methods than lists because they are immutable.

Two useful tuple methods are:

- `count(value)` returns the number of occurrences of a value.
- `index(value)` returns the index of the first occurrence of a value.

```python
numbers = (10, 20, 30, 20, 40)

print(numbers.count(20))
print(numbers.index(30))
```

## 25. List vs Tuple

Lists and tuples are both ordered collections that support indexing and duplicate values.

The main difference is mutability:

- Lists are mutable.
- Tuples are immutable.

Use a list when the collection needs to change.

Use a tuple when the collection should remain fixed.

```python
student_scores = [18, 15, 20]
coordinates = (35.7, 51.4)

student_scores[0] = 19

print(student_scores)
print(coordinates)
```

## 26. Introduction to Sets

A set is a collection of unique items.

Important characteristics of sets:

- Duplicate values are removed.
- Sets are mutable.
- Sets do not support indexing.
- Sets are useful for finding unique values and performing mathematical set operations.

```python
numbers = {10, 20, 30, 20, 10}

print(numbers)
print(type(numbers))
```

```python
empty_set = set()
empty_dictionary = {}

print(type(empty_set))
print(type(empty_dictionary))
```

## 27. Removing Duplicate Values

Sets can be used to remove duplicate values from a collection.

Converting a list to a set keeps only unique values.

However, the original order of the list is not guaranteed to be preserved.

```python
scores = [18, 15, 20, 18, 15, 12]

unique_scores = set(scores)

print(unique_scores)
```

## 28. Adding Items to a Set

The `add()` method adds one item to a set.

If the item already exists, the set remains unchanged.

```python
numbers = {10, 20, 30}

numbers.add(40)
numbers.add(20)

print(numbers)
```

## 29. Removing Items from a Set

Python provides several methods to remove items from a set.

- `remove(value)` removes an item and raises a `KeyError` if it does not exist.
- `discard(value)` removes an item without raising an error if it does not exist.
- `pop()` removes and returns an arbitrary item.
- `clear()` removes all items.

```python
numbers = {10, 20, 30, 40}

numbers.remove(20)
numbers.discard(50)

print(numbers)
```

```python
numbers = {10, 20, 30}

removed_item = numbers.pop()

print("Removed:", removed_item)
print("Remaining:", numbers)

numbers.clear()

print("After clear:", numbers)
```

## 30. Set Operations

Python supports common mathematical set operations:

- Union: combines all unique items from both sets.
- Intersection: returns items present in both sets.
- Difference: returns items present in the first set but not the second.
- Symmetric difference: returns items present in either set, but not both.

```python
gis_students = {"Ali", "Sara", "Reza", "Mina"}
python_students = {"Sara", "Reza", "Omid"}

print("Union:", gis_students | python_students)
print("Intersection:", gis_students & python_students)
print("Difference:", gis_students - python_students)
print("Symmetric difference:", gis_students ^ python_students)
```

## 31. Practice: Comparing Two Groups

Given two sets of students:

- Find all students who attend either course.
- Find students who attend both courses.
- Find students who attend GIS but not Python.
- Check whether `"Sara"` attends the Python course.

```python
gis_students = {"Ali", "Sara", "Reza", "Mina"}
python_students = {"Sara", "Reza", "Omid"}

all_students = gis_students | python_students
both_courses = gis_students & python_students
gis_only = gis_students - python_students
sara_attends_python = "Sara" in python_students

print("All students:", all_students)
print("Both courses:", both_courses)
print("GIS only:", gis_only)
print("Sara attends Python:", sara_attends_python)
```

## 32. Introduction to Dictionaries

A dictionary is a collection of key-value pairs.

Each key is associated with a value.

Important characteristics of dictionaries:

- Keys must be unique.
- Values can be duplicated.
- Dictionaries are mutable.
- Items are accessed using their keys.
- Dictionaries preserve insertion order.

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

print(student)
print(type(student))
```

## 33. Accessing Dictionary Values

We can access a dictionary value using its key.

If the key does not exist, square bracket access raises a `KeyError`.

The `get()` method provides a safer alternative when a key may be missing.

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

print(student["name"])
print(student["age"])
```

```python
student = {
    "name": "Barbod",
    "age": 23
}

print(student.get("name"))
print(student.get("major"))
print(student.get("major", "Unknown"))
```

## 34. Adding and Updating Dictionary Items

We can add a new key-value pair by assigning a value to a new key.

If the key already exists, its value is updated.

```python
student = {
    "name": "Barbod",
    "age": 23
}

student["major"] = "GIS"
student["age"] = 24

print(student)
```

## 35. Removing Dictionary Items

Python provides several ways to remove dictionary items.

- `pop(key)` removes the specified key and returns its value.
- `del` removes an item using its key.
- `popitem()` removes and returns the last inserted key-value pair.
- `clear()` removes all items.

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

removed_value = student.pop("age")

print("Removed value:", removed_value)
print("Student:", student)
```

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

del student["age"]

last_item = student.popitem()

print("Removed item:", last_item)
print("Student:", student)
```

## 36. Dictionary Methods: `keys()`, `values()`, and `items()`

Dictionaries provide three useful methods:

- `keys()` returns a view of the dictionary's keys.
- `values()` returns a view of its values.
- `items()` returns a view of its key-value pairs.

These views can be used to iterate through dictionary data.

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

print(student.keys())
print(student.values())
print(student.items())
```

## 37. Looping Through a Dictionary

A `for` loop iterates over dictionary keys by default.

We can use `values()` to iterate over values and `items()` to iterate over key-value pairs.

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

for key in student:
    print(key)
```

```python
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(key, ":", value)
```

## 38. Checking Keys and Dictionary Length

The `in` operator checks whether a key exists in a dictionary.

The `len()` function returns the number of key-value pairs.

```python
student = {
    "name": "Barbod",
    "age": 23,
    "major": "GIS"
}

print("name" in student)
print("Barbod" in student)
print(len(student))
```

```python
print("Barbod" in student.values())
```

## 39. Nested Dictionaries

A dictionary can contain other dictionaries as values.

This structure is useful for representing hierarchical or related data.

```python
students = {
    "student_1": {
        "name": "Ali",
        "score": 18
    },
    "student_2": {
        "name": "Sara",
        "score": 20
    }
}

print(students["student_1"]["name"])
print(students["student_2"]["score"])
```

## 41. Comparing Lists, Tuples, Sets, and Dictionaries

Python provides different data structures for different purposes.

- Use a list for an ordered collection that may change.
- Use a tuple for an ordered collection that should remain fixed.
- Use a set for unique items and set operations.
- Use a dictionary to associate keys with values.

```python
scores = [18, 15, 20]
coordinates = (35.7, 51.4)
unique_ids = {101, 102, 103}
student = {"name": "Ali", "score": 18}

print(scores)
print(coordinates)
print(unique_ids)
print(student)
```

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Ordered | Yes | Yes | No guaranteed order | Yes |
| Mutable | Yes | No | Yes | Yes |
| Allows duplicates | Yes | Yes | No | Keys must be unique |
| Access by index | Yes | Yes | No | Access by key |
| Main purpose | Changeable sequence | Fixed sequence | Unique items | Key-value mapping |

## 43. Final Practice: Combining Data Structures

You are given a list of student records.

Each record is a dictionary containing a student's name and scores.

Complete the following tasks:

1. Print each student's name.
2. Calculate each student's average score.
3. Find the names of students whose average score is at least 18.
4. Store those names in a set to ensure uniqueness.

```python
students = [
    {
        "name": "Ali",
        "scores": [18, 20, 19]
    },
    {
        "name": "Sara",
        "scores": [17, 18, 19]
    },
    {
        "name": "Reza",
        "scores": [14, 16, 15]
    },
    {
        "name": "Mina",
        "scores": [20, 19, 20]
    }
]
```

```python
high_performers = set()

for student in students:
    name = student["name"]
    scores = student["scores"]

    average_score = sum(scores) / len(scores)

    print(name, "Average:", average_score)

    if average_score >= 18:
        high_performers.add(name)

print("High performers:", high_performers)
```

## 44. Chapter Summary

In this chapter, we learned about Python's four main built-in data structures.

### Lists
- Ordered and mutable.
- Allow duplicate values.
- Support indexing, slicing, and many useful methods.

### Tuples
- Ordered and immutable.
- Allow duplicate values.
- Support indexing and tuple unpacking.
