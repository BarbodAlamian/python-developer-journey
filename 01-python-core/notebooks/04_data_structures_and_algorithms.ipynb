Choosing the right data structure can make a program easier to understand and more efficient.

## 2. What Is an Algorithm?

An **algorithm** is a sequence of steps used to solve a problem.

An algorithm receives some input, performs a series of operations, and produces an output.

For example, an algorithm for finding the largest number in a list could:

1. Assume the first number is the largest.
2. Examine each remaining number.
3. If a larger number is found, update the current largest number.
4. Return the largest number after examining all items.

Algorithms help us solve problems in a clear and systematic way.

````python
scores = [14, 18, 11, 20, 16]

largest = scores[0]

for score in scores[1:]:
    if score > largest:
        largest = score

print("Largest score:", largest)
````

## 3. Data Structures vs. Algorithms

Data structures and algorithms are closely related, but they are not the same.

- A **data structure** organizes and stores data.
- An **algorithm** defines the steps used to solve a problem.

For example, a list can store a collection of numbers, while a searching algorithm can find a specific number in that list.

The data structure determines how the data is organized, and the algorithm determines how we process it.

## 5. What Is Time Complexity?

**Time complexity** describes how the amount of work performed by an algorithm grows as the input size increases.

It does not measure the exact execution time in seconds. Instead, it helps us understand how an algorithm behaves when it processes more data.

We usually represent the input size with the letter `n`.

For example:

- If `n = 10`, the algorithm receives 10 items.
- If `n = 100`, the algorithm receives 100 items.
- If `n = 1,000`, the algorithm receives 1,000 items.

An efficient algorithm should ideally avoid doing unnecessary work as the input size grows.

Time complexity helps us compare algorithms and choose suitable solutions for larger problems.

## 6. Common Time Complexities

### 6.1. Constant Time — O(1)

An algorithm has **constant time complexity** when the amount of work does not grow with the input size.

For example, accessing an item at a known index in a Python list takes constant time.

### 6.2. Linear Time — O(n)

An algorithm has **linear time complexity** when its amount of work grows proportionally to the input size.

For example, printing every item in a list requires visiting each item once.

### 6.3. Quadratic Time — O(n²)

An algorithm has **quadratic time complexity** when its amount of work grows proportionally to the square of the input size.

A common example is a nested loop in which each iteration of the outer loop processes all items in the inner loop.

As the input grows, a quadratic algorithm can become much slower than a linear algorithm.

````python
numbers = [10, 20, 30, 40, 50]

for first in numbers:
    for second in numbers:
        print(first, second)
````

## 8. How to Analyze Big O

Big O notation describes how an algorithm's work grows as the input size increases.

When analyzing Big O, we focus on the dominant growth rate rather than the exact number of operations.

We use three useful rules:

1. **Ignore constant factors.** For example, `O(2n)` simplifies to `O(n)`.
2. **Keep the fastest-growing term.** For example, `O(n² + n)` simplifies to `O(n²)`.
3. **Analyze separate loops individually.** If two loops each process `n` items one after another, their combined complexity is `O(n + n)`, which simplifies to `O(n)`.

These rules help us describe an algorithm's growth in a simple and consistent way.

## 10. Linear Search

**Linear search** is an algorithm that searches for a target value by examining items one at a time.

The algorithm:

1. Starts at the first item.
2. Compares each item with the target.
3. Returns the index if the target is found.
4. Returns `-1` if the target does not exist in the list.

Linear search works with both sorted and unsorted lists.

Its worst-case time complexity is `O(n)` because it may need to examine every item.

````python
def linear_search(numbers, target):
    for index, number in enumerate(numbers):
        if number == target:
            return index

    return -1


numbers = [12, 5, 19, 3, 8]

print(linear_search(numbers, 19))
print(linear_search(numbers, 7))
````

## 12. Binary Search

**Binary search** is an algorithm that finds a target value in a sorted list by repeatedly dividing the search range in half.

Unlike linear search, binary search requires the list to be sorted.

The algorithm:

1. Set the left and right boundaries of the search range.
2. Find the middle index.
3. Compare the middle value with the target.
4. If they are equal, return the middle index.
5. If the target is smaller, continue searching the left half.
6. If the target is larger, continue searching the right half.
7. If the search range becomes empty, return `-1`.

Binary search has a worst-case time complexity of `O(log n)`.

````python
numbers = [3, 5, 8, 12, 16, 19, 24, 31, 42]
````

````python
def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = (left + right) // 2

        if numbers[middle] == target:
            return middle

        elif numbers[middle] < target:
            left = middle + 1

        else:
            right = middle - 1

    return -1


numbers = [3, 5, 8, 12, 16, 19, 24, 31, 42]

print(binary_search(numbers, 19))
print(binary_search(numbers, 7))
````

## 13. Linear Search vs. Binary Search

Linear search and binary search solve the same problem, but they use different strategies.

| Feature | Linear Search | Binary Search |
|---|---|---|
| Data must be sorted | No | Yes |
| Search strategy | Check items one by one | Repeatedly divide the range in half |
| Worst-case time complexity | `O(n)` | `O(log n)` |
| Best suited for | Small or unsorted lists | Sorted lists |

Binary search is generally much more efficient for searching large, sorted lists.

However, sorting an unsorted list takes time too. Therefore, the cost of sorting may matter if we only need to perform a single search.

## 16. Bubble Sort

**Bubble sort** is a sorting algorithm that repeatedly compares adjacent elements and swaps them if they are in the wrong order.

For ascending order:

1. Compare the first two elements.
2. Swap them if the first element is greater than the second.
3. Move to the next pair and repeat.
4. After one complete pass, the largest unsorted element reaches the end.
5. Repeat the process until the list is sorted.

Bubble sort has an average and worst-case time complexity of `O(n²)`.

It is useful for learning sorting concepts, but it is generally inefficient for large datasets.

````python
def bubble_sort(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        for j in range(n - 1 - i):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = (
                    numbers[j + 1],
                    numbers[j]
                )

    return numbers


numbers = [5, 2, 8, 1, 3]

sorted_numbers = bubble_sort(numbers)

print("Original list:", numbers)
print("Sorted list:", sorted_numbers)
````

## 18. Selection Sort

**Selection sort** repeatedly finds the smallest element in the unsorted portion of a list and moves it to the beginning of that portion.

The algorithm:

1. Start at the first position.
2. Find the smallest element in the remaining unsorted portion.
3. Swap it with the element at the current position.
4. Move to the next position.
5. Repeat until the list is sorted.

Selection sort has `O(n²)` time complexity in the average and worst cases.

Unlike bubble sort, selection sort usually performs at most one swap per outer-loop iteration.

````python
def selection_sort(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if numbers[j] < numbers[min_index]:
                min_index = j

        numbers[i], numbers[min_index] = (
            numbers[min_index],
            numbers[i]
        )

    return numbers


numbers = [5, 2, 8, 1, 3]

print(selection_sort(numbers))
````

## 19. Bubble Sort vs. Selection Sort

Both bubble sort and selection sort have `O(n²)` average and worst-case time complexity.

Their main difference is how they organize the data:

- **Bubble sort:** Repeatedly compares adjacent elements and swaps them when necessary.
- **Selection sort:** Finds the smallest remaining element and places it in the next sorted position.

Both algorithms are useful for learning, but Python's built-in sorting tools are generally a better choice in real applications.

## 20. Insertion Sort

**Insertion sort** builds a sorted portion of a list one element at a time.

It works similarly to arranging playing cards in your hand.

The algorithm:

1. Treat the first element as the initially sorted portion.
2. Take the next element as the current value.
3. Compare it with the elements in the sorted portion, moving larger elements one position to the right.
4. Insert the current value into its correct position.
5. Repeat until every element has been processed.

Insertion sort has `O(n²)` average and worst-case time complexity. Its best-case time complexity is `O(n)` when the list is already sorted.

It can be effective for small lists or lists that are already nearly sorted.

````python
def insertion_sort(numbers):
    numbers = numbers.copy()

    for i in range(1, len(numbers)):
        current = numbers[i]
        j = i - 1

        while j >= 0 and numbers[j] > current:
            numbers[j + 1] = numbers[j]
            j -= 1

        numbers[j + 1] = current

    return numbers


numbers = [5, 2, 4, 1, 3]

print("Original list:", numbers)
print("Sorted list:", insertion_sort(numbers))
````

## 21. Comparing Sorting Algorithms

| Algorithm | Main Idea | Best Case | Average Case | Worst Case |
|---|---|---|---|---|
| Bubble Sort | Swap adjacent elements | `O(n)` with early stopping | `O(n²)` | `O(n²)` |
| Selection Sort | Find the smallest remaining element | `O(n²)` | `O(n²)` | `O(n²)` |
| Insertion Sort | Insert each element into its correct position | `O(n)` | `O(n²)` | `O(n²)` |

The best-case complexity of bubble sort shown above assumes an optimized implementation that stops when no swaps occur during a pass.

All three algorithms are useful for learning, but they are generally not the preferred choice for sorting large datasets in production applications.

````python
def bubble_sort_optimized(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        swapped = False

        for j in range(n - 1 - i):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = (
                    numbers[j + 1],
                    numbers[j]
                )
                swapped = True

        if not swapped:
            break

    return numbers


print(bubble_sort_optimized([1, 2, 3, 4, 5]))
print(bubble_sort_optimized([5, 2, 4, 1, 3]))
````

## 24. Introduction to Recursion

**Recursion** is a programming technique in which a function calls itself to solve a smaller version of the same problem.

A recursive function needs two important parts:

1. **Base case:** The condition that stops the recursion.
2. **Recursive case:** The part where the function calls itself with a smaller or simpler input.

Without a correct base case, a recursive function may continue calling itself until Python raises a `RecursionError`.

Recursion is useful for problems that can naturally be divided into smaller subproblems.

````python
def countdown(number):
    if number == 0:
        print("Done!")
        return

    print(number)
    countdown(number - 1)


countdown(3)
````

## 25. Factorial Using Recursion

The factorial of a non-negative integer `n`, written as `n!`, is the product of all positive integers from `1` to `n`.

For example:

`4! = 4 × 3 × 2 × 1 = 24`

The factorial can be defined recursively:

- `0! = 1` — the base case.
- `n! = n × (n - 1)!` for `n > 0` — the recursive case.

This definition naturally translates into a recursive Python function.

````python
def factorial(number):
    if number < 0:
        raise ValueError("Factorial is not defined for negative integers.")

    if number == 0:
        return 1

    return number * factorial(number - 1)


print(factorial(4))
print(factorial(5))
print(factorial(0))
````

## 26. Recursion vs. Iteration

**Iteration** means repeating operations using loops such as `for` and `while`.

**Recursion** means solving a problem through repeated function calls, where each call works on a smaller or simpler problem.

Both approaches can solve many of the same problems.

| Feature | Iteration | Recursion |
|---|---|---|
| Main mechanism | Loops | Function calls |
| Stopping condition | Loop condition | Base case |
| Memory usage | Often lower for simple repetition | Uses additional call-stack space |
| Common risk | Infinite loops | Excessive recursion depth |

Recursion can make some algorithms easier to express, but it is not always the most efficient solution.

````python
def factorial_iterative(number):
    if number < 0:
        raise ValueError("Factorial is not defined for negative integers.")

    result = 1

    for value in range(1, number + 1):
        result *= value

    return result


def factorial_recursive(number):
    if number < 0:
        raise ValueError("Factorial is not defined for negative integers.")

    if number == 0:
        return 1

    return number * factorial_recursive(number - 1)


print(factorial_iterative(5))
print(factorial_recursive(5))
````

## 28. Stack

A **stack** is a data structure that follows the **Last In, First Out (LIFO)** principle.

This means that the last item added to the stack is the first item removed.

Imagine a stack of plates: you place a new plate on top, and you usually remove the top plate first.

The main stack operations are:

- **Push:** Add an item to the top.
- **Pop:** Remove and return the top item.
- **Peek:** View the top item without removing it.
- **is_empty:** Check whether the stack contains any items.

In Python, a list can be used to implement a simple stack with `append()` and `pop()`.

````python
stack = []

stack.append("Page 1")
stack.append("Page 2")
stack.append("Page 3")

print("Stack:", stack)

last_item = stack.pop()

print("Removed item:", last_item)
print("Stack after pop:", stack)
````

````python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Cannot pop from an empty stack.")

        return self._items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek into an empty stack.")

        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def __len__(self):
        return len(self._items)


stack = Stack()

stack.push("A")
stack.push("B")
stack.push("C")

print("Top item:", stack.peek())
print("Removed:", stack.pop())
print("Stack size:", len(stack))
print("Is empty:", stack.is_empty())
````

## 29. Queue

A **queue** is a data structure that follows the **First In, First Out (FIFO)** principle.

This means that the first item added to the queue is the first item removed.

Imagine a line of people waiting to enter a store: the person who arrives first is served first.

The main queue operations are:

- **Enqueue:** Add an item to the back of the queue.
- **Dequeue:** Remove and return the item at the front.
- **Peek:** View the front item without removing it.
- **is_empty:** Check whether the queue contains any items.

In Python, `collections.deque` is a suitable choice for implementing a queue because adding and removing items at either end are efficient operations.

````python
from collections import deque

queue = deque()

queue.append("Customer 1")
queue.append("Customer 2")
queue.append("Customer 3")

print("Queue:", queue)

served_customer = queue.popleft()

print("Served:", served_customer)
print("Queue after dequeue:", queue)
````

````python
from collections import deque


class Queue:
    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        self._items.append(item)

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Cannot dequeue from an empty queue.")

        return self._items.popleft()

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek into an empty queue.")

        return self._items[0]

    def is_empty(self):
        return len(self._items) == 0

    def __len__(self):
        return len(self._items)


queue = Queue()

queue.enqueue("Task 1")
queue.enqueue("Task 2")
queue.enqueue("Task 3")

print("Front item:", queue.peek())
print("Processed:", queue.dequeue())
print("Queue size:", len(queue))
print("Is empty:", queue.is_empty())
````

## 32. Introduction to Linked Lists

A **linked list** is a data structure made up of individual elements called **nodes**.

Each node in a singly linked list contains two things:

- **Data:** The value stored in the node.
- **Next:** A reference to the next node in the list.

The first node is called the **head**. The last node points to `None`, indicating that there are no more nodes.

Unlike a Python list, whose elements can be accessed directly by index, a linked list is typically traversed from node to node.

Linked lists are useful for learning how references connect objects and how data structures can be built from smaller components.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


first = Node(10)
second = Node(20)
third = Node(30)

first.next = second
second.next = third

print(first.data)
print(first.next.data)
print(first.next.next.data)
print(first.next.next.next)
````

## 43. Introduction to Heaps

## 53. Breadth-First Search (BFS)

**Breadth-First Search (BFS)** explores a graph level by level, starting from a chosen vertex.

BFS uses a queue to keep track of vertices waiting to be explored.

A `visited` set prevents the algorithm from visiting the same vertex repeatedly.

For an adjacency-list graph, BFS has a time complexity of `O(V + E)`, where:
- `V` is the number of vertices.
- `E` is the number of edges.

````python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

bfs(graph, "A")
````

## 54. Depth-First Search (DFS)

**Depth-First Search (DFS)** explores one path as deeply as possible before backtracking.

DFS can be implemented using:
- Recursion.
- An explicit stack.

A `visited` set prevents repeated visits and infinite traversal in cyclic graphs.

For an adjacency-list graph, DFS has a time complexity of `O(V + E)`.

````python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

dfs(graph, "A")
````

## 58. Introduction to Dijkstra's Algorithm

**Dijkstra's Algorithm** finds the shortest paths from a starting vertex to other vertices in a weighted graph, provided that all edge weights are non-negative.

A **weighted graph** is a graph in which each edge has a numerical value, such as distance, travel time, or cost.

Dijkstra's Algorithm repeatedly selects the unvisited vertex with the smallest known distance and updates the distances to its neighbors.

````python
graph = {
    "A": {"B": 4, "C": 2},
    "B": {"A": 4, "C": 1, "D": 5},
    "C": {"A": 2, "B": 1, "D": 8},
    "D": {"B": 5, "C": 8}
}

print(graph["A"])
print(graph["B"]["D"])
````

## 60. Dijkstra's Algorithm: Core Idea

Dijkstra's Algorithm starts by assigning:
- Distance `0` to the starting vertex.
- Distance `infinity` to every other vertex.

At each step, it selects the unvisited vertex with the smallest known distance.

It then checks each neighbor and asks:

**Can I reach this neighbor more cheaply through the current vertex?**

If the new route is shorter, the neighbor's distance is updated.

The process continues until all reachable vertices have been visited.

````python
def dijkstra(graph, start):
    distances = {node: float("inf") for node in graph}
    distances[start] = 0

    visited = set()

    while len(visited) < len(graph):
        current = min(
            (
                node for node in graph
                if node not in visited
            ),
            key=lambda node: distances[node],
            default=None
        )

        if current is None or distances[current] == float("inf"):
            break

        visited.add(current)

        for neighbor, weight in graph[current].items():
            if neighbor in visited:
                continue

            new_distance = distances[current] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance

    return distances


graph = {
    "A": {"B": 4, "C": 2},
    "B": {"A": 4, "C": 1, "D": 5},
    "C": {"A": 2, "B": 1, "D": 8},
    "D": {"B": 5, "C": 8}
}

print(dijkstra(graph, "A"))
````

## 61. Understanding Dijkstra's Algorithm

Dijkstra's algorithm finds the shortest distance from a starting node to every other reachable node in a weighted graph.

The algorithm:
1. Sets the starting node's distance to `0`.
2. Sets all other distances to infinity.
3. Selects the unvisited node with the smallest known distance.
4. Updates the distances of its neighbors when a shorter path is found.
5. Repeats until all reachable nodes have been visited.

Dijkstra's algorithm requires non-negative edge weights.

````python
print(dijkstra(graph, "A"))
print(dijkstra(graph, "B"))
print(dijkstra(graph, "C"))
print(dijkstra(graph, "D"))
````

## 64. Introduction to A* Search

A* is a pathfinding algorithm that searches for a low-cost path from a start node to a goal node.

It evaluates each node using:

f(n) = g(n) + h(n)

Where:

- `g(n)` is the actual cost from the start node to node `n`.
- `h(n)` is the estimated cost from node `n` to the goal.
- `f(n)` is the estimated total cost of a path through node `n`.

Dijkstra's algorithm prioritizes the smallest known distance from the start. A* also considers an estimate of the remaining distance to the goal.

For A* to guarantee an optimal path, its heuristic must not overestimate the true remaining cost. This is called an admissible heuristic.

## 65. Defining a Heuristic

A heuristic estimates the remaining cost from a node to the goal.

In this example, the goal is node `D`.

We define a heuristic value for each node:

- `A`: Estimated cost to reach `D` is `6`.
- `B`: Estimated cost to reach `D` is `5`.
- `C`: Estimated cost to reach `D` is `5`.
- `D`: Estimated cost to reach itself is `0`.

A heuristic is admissible if it never overestimates the actual minimum cost to reach the goal.

These values are admissible for our graph because they do not exceed the actual shortest distances to `D`.

````python
heuristic = {
    "A": 6,
    "B": 5,
    "C": 5,
    "D": 0
}

print(heuristic["A"])
print(heuristic["D"])
````

## 66. Implementing A* Search

We will implement A* using Python's built-in `heapq` module.

The priority queue stores three values:

- The estimated total cost, `f(n)`.
- The actual distance from the start, `g(n)`.
- The current node.

The algorithm updates a node's distance when it discovers a cheaper path. It also stores each node's predecessor so that the final path can be reconstructed.

The function returns the shortest path and its total cost. If no path exists, it returns an empty path and infinity.

````python
import heapq


def a_star(graph, start, goal, heuristic):
    distances = {node: float("inf") for node in graph}
    distances[start] = 0

    previous = {}
    priority_queue = [(heuristic[start], 0, start)]

    while priority_queue:
        _, current_distance, current = heapq.heappop(priority_queue)

        if current_distance != distances[current]:
            continue

        if current == goal:
            path = []
            while current in previous:
                path.append(current)
                current = previous[current]

            path.append(start)
            path.reverse()

            return path, distances[goal]

        for neighbor, weight in graph[current].items():
            new_distance = current_distance + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current

                priority = new_distance + heuristic[neighbor]

                heapq.heappush(
                    priority_queue,
                    (priority, new_distance, neighbor)
                )

    return [], float("inf")
````

## 67. Testing A* Search

We will find the shortest path from node `A` to node `D`.

The function returns two values:

- `path`: The sequence of nodes in the selected path.
- `distance`: The total cost of that path.

We expect the path `A → C → B → D`, with a total cost of `8`.

````python
path, distance = a_star(graph, "A", "D", heuristic)

print("Shortest path:", path)
print("Total cost:", distance)
````

## 68. Understanding A* Step by Step

Starting from node `A`:

1. The cost to reach `A` is `0`.
2. From `A`, the algorithm considers `B` and `C`.
3. It calculates `f(n) = g(n) + h(n)` for each candidate.
4. It prioritizes the candidate with the smallest estimated total cost.
5. It updates distances and predecessors when a cheaper path is found.
6. When the goal is reached with its best known cost, the algorithm reconstructs the path.

A* uses the heuristic to guide the search toward the goal. The quality of the heuristic can affect how many nodes the algorithm explores.

## 69. Introduction to Greedy Algorithms

A greedy algorithm makes the best-looking choice at each step.

It chooses a locally optimal option without reconsidering previous decisions.

Greedy algorithms can be efficient and easy to implement. However, a locally optimal choice does not always lead to a globally optimal solution.

Examples of problems that can be solved using greedy approaches include:
- Activity selection
- Fractional knapsack
- Minimum spanning tree algorithms such as Kruskal's and Prim's algorithms

A greedy strategy is appropriate when the problem has the properties needed to guarantee that its local choices lead to an optimal solution.

## 70. A Simple Greedy Example

Suppose we want to select the largest possible number of non-overlapping activities.

Each activity has a start time and a finish time.

A greedy strategy is to:
1. Sort activities by finish time.
2. Select the activity that finishes earliest.
3. Select the next activity whose start time is at or after the finish time of the last selected activity.
4. Repeat until no activities remain.

Choosing the earliest-finishing activity leaves as much time as possible for the remaining activities.

````python
activities = [
    (5, 9),
    (1, 8),
    (8, 10),
    (2, 5),
    (1, 3),
    (4, 7)
]

sorted_activities = sorted(activities, key=lambda activity: activity[1])

print("Original activities:", activities)
print("Activities sorted by finish time:", sorted_activities)
````

## 71. Implementing Activity Selection

We will implement the activity selection algorithm as a Python function.

The function:
1. Sorts activities by their finish times.
2. Selects the first activity.
3. Checks each remaining activity.
4. Selects an activity if it starts at or after the last selected activity finishes.
5. Returns the selected activities.

This greedy strategy produces a maximum-size set of non-overlapping activities for the standard activity selection problem.

````python
def activity_selection(activities):
    activities = sorted(activities, key=lambda activity: activity[1])

    selected = []
    last_finish_time = float("-inf")

    for start, finish in activities:
        if start >= last_finish_time:
            selected.append((start, finish))
            last_finish_time = finish

    return selected
````

## 72. Testing Activity Selection

We will test the greedy strategy with a list of activities.

Each activity is represented by a tuple containing its start time and finish time.

The algorithm should select a maximum-size set of mutually compatible activities for this standard activity-selection problem.

````python
activities = [
    (1, 3),
    (2, 5),
    (4, 7),
    (1, 8),
    (5, 9),
    (8, 10)
]

selected = activity_selection(activities)

print("Selected activities:", selected)
print("Number of activities:", len(selected))
````

## 73. Introduction to Dynamic Programming

Dynamic Programming (DP) is a problem-solving technique that breaks a problem into smaller subproblems and stores their results to avoid solving the same subproblems repeatedly.

Dynamic Programming is especially useful when a problem has:

- **Overlapping subproblems:** The same smaller problems appear multiple times.
- **Optimal substructure:** An optimal solution can be built from optimal solutions to smaller subproblems.

There are two common approaches:

- **Memoization (Top-Down):** Uses recursion and stores computed results.
- **Tabulation (Bottom-Up):** Solves smaller subproblems first and stores their results in a table.

Dynamic Programming can improve efficiency by avoiding repeated calculations.

## 74. Fibonacci: The Recursive Approach

The Fibonacci sequence is defined as:

- `F(0) = 0`
- `F(1) = 1`
- `F(n) = F(n - 1) + F(n - 2)` for `n >= 2`

A simple recursive function follows this definition directly.

However, this approach repeatedly calculates the same values. As a result, its running time grows exponentially.

````python
def fibonacci_recursive(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n <= 1:
        return n

    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)


print(fibonacci_recursive(6))
print(fibonacci_recursive(10))
````

## 75. Fibonacci with Memoization

Memoization is a top-down Dynamic Programming technique.

It stores the result of each computed subproblem so that the same calculation does not need to be repeated.

In this example, we use a dictionary called `memo` to store previously calculated Fibonacci numbers.

For the standard Fibonacci problem, memoization reduces the time complexity from exponential time to O(n).

````python
def fibonacci_memoization(n, memo=None):
    if n < 0:
        raise ValueError("n must be non-negative")

    if memo is None:
        memo = {}

    if n <= 1:
        return n

    if n in memo:
        return memo[n]

    memo[n] = (
        fibonacci_memoization(n - 1, memo)
        + fibonacci_memoization(n - 2, memo)
    )

    return memo[n]


print(fibonacci_memoization(6))
print(fibonacci_memoization(10))
print(fibonacci_memoization(50))
````

## 76. Fibonacci with Tabulation

Tabulation is a bottom-up Dynamic Programming technique.

Instead of using recursion, it starts with the smallest subproblems and builds up to the required result.

In this implementation:
1. We create a list to store Fibonacci numbers.
2. We initialize the first two values.
3. We calculate each subsequent value using the previous two values.
4. We return the value at index `n`.

This approach runs in O(n) time and uses O(n) space.

````python
def fibonacci_tabulation(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n <= 1:
        return n

    dp = [0] * (n + 1)
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


print(fibonacci_tabulation(6))
print(fibonacci_tabulation(10))
print(fibonacci_tabulation(50))
````

## 77. Comparing Fibonacci Approaches

We have implemented three approaches to calculate Fibonacci numbers:

- **Recursion:** Simple and closely follows the mathematical definition, but repeats many calculations.
- **Memoization:** Uses recursion and caches previously computed results.
- **Tabulation:** Builds the solution iteratively from smaller subproblems.

Both memoization and tabulation run in O(n) time for this problem.

Memoization uses O(n) additional space for cached results and the recursion stack. Tabulation uses O(n) space for its list.

The recursive approach without caching has exponential time complexity.

````python
n = 10

print("Recursive:", fibonacci_recursive(n))
print("Memoization:", fibonacci_memoization(n))
print("Tabulation:", fibonacci_tabulation(n))
````

## 78. The Climbing Stairs Problem

Suppose you are climbing a staircase with `n` steps.

At each move, you can climb either one step or two steps.

The goal is to determine how many distinct ways there are to reach the top.

For example, if there are three steps, the possible ways are:

- `1 + 1 + 1`
- `1 + 2`
- `2 + 1`

Therefore, there are three distinct ways to climb three steps.

This problem can be solved using Dynamic Programming because the number of ways to reach a step depends on the results for smaller steps.

## 79. Climbing Stairs: Recursive Approach

At each step, we can climb either one or two stairs.

To reach step `n`, we must have come from either:
- Step `n - 1`, by taking one step.
- Step `n - 2`, by taking two steps.

Therefore, the recurrence relation is:

`ways(n) = ways(n - 1) + ways(n - 2)`

The base cases are:
- `ways(0) = 1`: There is one way to stay at the ground level.
- `ways(1) = 1`: There is one way to climb one step.

This recursive solution is easy to understand, but it repeats calculations and has exponential time complexity.

````python
def climbing_stairs_recursive(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n <= 1:
        return 1

    return (
        climbing_stairs_recursive(n - 1)
        + climbing_stairs_recursive(n - 2)
    )


print(climbing_stairs_recursive(2))
print(climbing_stairs_recursive(3))
print(climbing_stairs_recursive(5))
````

## 80. Climbing Stairs with Memoization

Memoization stores the results of previously solved subproblems.

When the function needs the number of ways to reach a step that has already been calculated, it reuses the stored result instead of repeating the calculation.

This reduces the time complexity from exponential time to O(n).

The dictionary stores the number of ways to reach each step.

````python
def climbing_stairs_memoization(n, memo=None):
    if n < 0:
        raise ValueError("n must be non-negative")

    if memo is None:
        memo = {}

    if n <= 1:
        return 1

    if n in memo:
        return memo[n]

    memo[n] = (
        climbing_stairs_memoization(n - 1, memo)
        + climbing_stairs_memoization(n - 2, memo)
    )

    return memo[n]


print(climbing_stairs_memoization(5))
print(climbing_stairs_memoization(10))
print(climbing_stairs_memoization(30))
````

## 81. Climbing Stairs with Tabulation

Tabulation solves the problem from the bottom up.

We start by initializing the number of ways to reach steps `0` and `1`.

Then, for each subsequent step, we add the number of ways to reach the previous two steps.

The resulting time complexity is O(n), and the space complexity is O(n) because we store the results in a list.

````python
def climbing_stairs_tabulation(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    if n <= 1:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


print(climbing_stairs_tabulation(5))
print(climbing_stairs_tabulation(10))
print(climbing_stairs_tabulation(30))
````

## 82. Space-Optimized Dynamic Programming

The tabulation solution stores every result in a list.

However, each new result depends only on the previous two results.

Therefore, we can store only two variables instead of the entire list.

This reduces the space complexity from O(n) to O(1), while keeping the time complexity at O(n).

````python
def climbing_stairs_optimized(n):
    if n < 0:
        raise ValueError("n must be non-negative")

    previous_two = 1
    previous_one = 1

    for _ in range(2, n + 1):
        current = previous_two + previous_one
        previous_two = previous_one
        previous_one = current

    return previous_one


print(climbing_stairs_optimized(5))
print(climbing_stairs_optimized(10))
print(climbing_stairs_optimized(30))
````

## 83. Introduction to the Coin Change Problem

The Coin Change problem involves a set of coin denominations and a target amount.

One common version asks for the minimum number of coins needed to make the target amount.

For example, given coin denominations `[1, 3, 4]` and a target amount of `6`, we can use two coins:

- `3 + 3 = 6`

The minimum number of coins is therefore `2`.

A greedy algorithm that repeatedly chooses the largest available coin does not always find the optimal solution.

Dynamic Programming can solve this problem by considering smaller amounts and building up to the target amount.

## 84. Coin Change: Greedy Approach

The greedy approach repeatedly chooses the largest coin that does not exceed the remaining amount.

This strategy is simple, but it does not always produce the minimum number of coins.

We will use coin denominations `[1, 3, 4]` and target amount `6`.

The greedy algorithm chooses `4`, then `1`, then another `1`, using three coins.

However, the optimal solution is `3 + 3`, which uses only two coins.

````python
def coin_change_greedy(coins, amount):
    if amount < 0:
        raise ValueError("amount must be non-negative")

    if any(coin <= 0 for coin in coins):
        raise ValueError("coin denominations must be positive")

    coins = sorted(coins, reverse=True)
    selected_coins = []

    for coin in coins:
        while amount >= coin:
            amount -= coin
            selected_coins.append(coin)

    if amount != 0:
        return []

    return selected_coins


print(coin_change_greedy([1, 3, 4], 6))
print(len(coin_change_greedy([1, 3, 4], 6)))
````

## 85. Coin Change with Dynamic Programming

We can solve the minimum coin change problem using Dynamic Programming.

Let `dp[amount]` represent the minimum number of coins needed to make that amount.

We initialize:
- `dp[0] = 0`, because no coins are needed to make amount zero.
- All other values are initialized to infinity because their minimum coin counts are not known yet.

For each amount, we consider every available coin. If using a coin leads to a smaller number of coins, we update the result.

The recurrence relation is:

`dp[amount] = min(dp[amount], dp[amount - coin] + 1)`

We process amounts from smallest to largest, so the smaller results are already available.

The time complexity is O(A × C), where `A` is the target amount and `C` is the number of coin denominations.

````python
def coin_change_dp(coins, amount):
    if amount < 0:
        raise ValueError("amount must be non-negative")

    if any(coin <= 0 for coin in coins):
        raise ValueError("coin denominations must be positive")

    dp = [float("inf")] * (amount + 1)
    dp[0] = 0

    for current_amount in range(1, amount + 1):
        for coin in coins:
            if coin <= current_amount:
                dp[current_amount] = min(
                    dp[current_amount],
                    dp[current_amount - coin] + 1
                )

    if dp[amount] == float("inf"):
        return -1

    return dp[amount]


print(coin_change_dp([1, 3, 4], 6))
print(coin_change_dp([1, 3, 4], 7))
print(coin_change_dp([1, 2, 5], 11))
````

## 86. Reconstructing the Selected Coins

So far, our Dynamic Programming function returns only the minimum number of coins.

Sometimes, we also need to know which coins form the solution.

We can store the last coin used for each amount. After calculating the minimum number of coins, we trace these choices backward until we reach zero.

This allows us to return both the minimum coin count and the selected coins.

````python
def coin_change_solution(coins, amount):
    if amount < 0:
        raise ValueError("amount must be non-negative")

    if any(coin <= 0 for coin in coins):
        raise ValueError("coin denominations must be positive")

    dp = [float("inf")] * (amount + 1)
    last_coin = [None] * (amount + 1)
    dp[0] = 0

    for current_amount in range(1, amount + 1):
        for coin in coins:
            if coin <= current_amount:
                candidate = dp[current_amount - coin] + 1

                if candidate < dp[current_amount]:
                    dp[current_amount] = candidate
                    last_coin[current_amount] = coin

    if dp[amount] == float("inf"):
        return -1, []

    selected_coins = []
    current_amount = amount

    while current_amount > 0:
        coin = last_coin[current_amount]
        selected_coins.append(coin)
        current_amount -= coin

    return dp[amount], selected_coins


count, selected = coin_change_solution([1, 3, 4], 6)

print("Minimum number of coins:", count)
print("Selected coins:", selected)
print("Total amount:", sum(selected))
````

## 87. Handling Impossible Amounts

Some target amounts cannot be formed from the available coin denominations.

For example, if the only available coin has a denomination of `2`, it is impossible to make an amount of `3`.

Our Dynamic Programming implementation returns `-1` when no valid solution exists.

An amount of zero is a valid special case: it requires zero coins.

````python
print(coin_change_dp([2], 3))
print(coin_change_dp([2], 4))
print(coin_change_dp([1, 3, 4], 0))
````

## 88. Greedy vs. Dynamic Programming

Greedy and Dynamic Programming are different problem-solving strategies.

**Greedy:**
- Makes a locally optimal choice at each step.
- Can be efficient and simple.
- Does not always guarantee an optimal solution.

**Dynamic Programming:**
- Solves and stores smaller subproblems.
- Reuses previously computed results.
- Can guarantee an optimal solution when the recurrence correctly represents the problem.

For coin denominations `[1, 3, 4]` and target amount `6`, Greedy uses three coins, while Dynamic Programming finds a solution using two coins.

This example demonstrates why a greedy strategy is not suitable for every optimization problem.

````python
coins = [1, 3, 4]
amount = 6

greedy_solution = coin_change_greedy(coins, amount)
dp_count = coin_change_dp(coins, amount)

print("Greedy coin count:", len(greedy_solution))
print("Dynamic Programming coin count:", dp_count)
````

## 89. Introduction to the 0/1 Knapsack Problem

The 0/1 Knapsack Problem is a classic optimization problem.

We are given:
- A set of items.
- A weight and a value for each item.
- A knapsack with a limited weight capacity.

The goal is to maximize the total value of the selected items without exceeding the capacity.

Each item has only two choices:
- Include the item.
- Exclude the item.

We cannot take a fraction of an item. This is why the problem is called 0/1 Knapsack.

Dynamic Programming can solve this problem by considering smaller subproblems involving items and capacities.

## 90. Defining the Knapsack Example

Consider three items with the following weights and values:

- Item 1: Weight `1`, Value `6`
- Item 2: Weight `2`, Value `10`
- Item 3: Weight `3`, Value `12`

The knapsack capacity is `5`.

We need to choose a combination of items whose total weight does not exceed `5` while maximizing the total value.

We will use this example to implement and test the 0/1 Knapsack algorithm.

````python
weights = [1, 2, 3]
values = [6, 10, 12]
capacity = 5

print("Weights:", weights)
print("Values:", values)
print("Capacity:", capacity)
````

## 91. Understanding the Knapsack Recurrence

For each item, we have two choices: include it or exclude it.

Let `dp[i][w]` represent the maximum value achievable using the first `i` items with a capacity of `w`.

If the current item is too heavy, we cannot include it:

`dp[i][w] = dp[i - 1][w]`

Otherwise, we choose the better of two options:

- Exclude the item: `dp[i - 1][w]`
- Include the item: `values[i - 1] + dp[i - 1][w - weights[i - 1]]`

Therefore, the recurrence is:

`dp[i][w] = max(dp[i - 1][w], values[i - 1] + dp[i - 1][w - weights[i - 1]])`

We initialize the first row and first column to zero because no items or zero capacity cannot produce any value.

````python
item_weight = 3
item_value = 12
remaining_capacity = 4
previous_best_value = 10

include_value = item_value + previous_best_value
exclude_value = 16

print("Include value:", include_value)
print("Exclude value:", exclude_value)
print("Best value:", max(include_value, exclude_value))
````

## 92. Implementing 0/1 Knapsack with Dynamic Programming

We will implement the 0/1 Knapsack algorithm using a two-dimensional table.

The rows represent the number of available items, and the columns represent the knapsack capacity.

For each item and capacity, we compare the values obtained by including and excluding the item.

Each row uses the previous row, ensuring that an item cannot be selected more than once.

The time complexity is O(n × W), and the space complexity is O(n × W), where `n` is the number of items and `W` is the capacity.

````python
def knapsack_01(weights, values, capacity):
    if capacity < 0:
        raise ValueError("capacity must be non-negative")

    if len(weights) != len(values):
        raise ValueError("weights and values must have equal lengths")

    if any(weight <= 0 for weight in weights):
        raise ValueError("weights must be positive")

    if any(value < 0 for value in values):
        raise ValueError("values must be non-negative")

    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight = weights[i - 1]
        value = values[i - 1]

        for w in range(capacity + 1):
            dp[i][w] = dp[i - 1][w]

            if weight <= w:
                include_value = value + dp[i - 1][w - weight]
                dp[i][w] = max(dp[i][w], include_value)

    return dp[n][capacity]


print(knapsack_01(weights, values, capacity))
````

## 93. Testing Different Capacities

We will test the same items with different knapsack capacities.

As capacity increases, the maximum achievable value cannot decrease because the previously available selections remain valid.

The algorithm should also return zero when the capacity is zero or when no items are available.

````python
print("Capacity 0:", knapsack_01(weights, values, 0))
print("Capacity 1:", knapsack_01(weights, values, 1))
print("Capacity 3:", knapsack_01(weights, values, 3))
print("Capacity 5:", knapsack_01(weights, values, 5))
print("No items:", knapsack_01([], [], 5))
````

## 94. Reconstructing the Selected Items

The previous implementation returns only the maximum achievable value.

Sometimes, we also need to know which items produce that value.

We can reconstruct the selected items by tracing backward through the Dynamic Programming table.

If the current table value differs from the value in the previous row, the current item was selected. We then reduce the remaining capacity by that item's weight.

The function will return the maximum value and the indices of the selected items.

````python
def knapsack_01_solution(weights, values, capacity):
    if capacity < 0:
        raise ValueError("capacity must be non-negative")

    if len(weights) != len(values):
        raise ValueError("weights and values must have equal lengths")

    if any(weight <= 0 for weight in weights):
        raise ValueError("weights must be positive")

    if any(value < 0 for value in values):
        raise ValueError("values must be non-negative")

    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight = weights[i - 1]
        value = values[i - 1]

        for w in range(capacity + 1):
            dp[i][w] = dp[i - 1][w]

            if weight <= w:
                include_value = value + dp[i - 1][w - weight]
                dp[i][w] = max(dp[i][w], include_value)

    selected_indices = []
    remaining_capacity = capacity

    for i in range(n, 0, -1):
        if dp[i][remaining_capacity] != dp[i - 1][remaining_capacity]:
            selected_indices.append(i - 1)
            remaining_capacity -= weights[i - 1]

    selected_indices.reverse()

    return dp[n][capacity], selected_indices


max_value, selected_indices = knapsack_01_solution(
    weights, values, capacity
)

print("Maximum value:", max_value)
print("Selected item indices:", selected_indices)
print("Selected weights:", [weights[i] for i in selected_indices])
print("Selected values:", [values[i] for i in selected_indices])
print("Total weight:", sum(weights[i] for i in selected_indices))
````

## 95. Space-Optimized 0/1 Knapsack

The two-dimensional implementation stores results for every item and capacity.

However, each row depends only on the previous row. We can therefore reduce the space complexity to O(W) by using a one-dimensional list.

The capacity loop must run backward, from the current capacity down to the item's weight.

This backward iteration is important: it prevents the same item from being used multiple times during a single iteration.

The optimized implementation keeps the O(n × W) time complexity but reduces the auxiliary space complexity to O(W).

````python
def knapsack_01_optimized(weights, values, capacity):
    if capacity < 0:
        raise ValueError("capacity must be non-negative")

    if len(weights) != len(values):
        raise ValueError("weights and values must have equal lengths")

    if any(weight <= 0 for weight in weights):
        raise ValueError("weights must be positive")

    if any(value < 0 for value in values):
        raise ValueError("values must be non-negative")

    dp = [0] * (capacity + 1)

    for weight, value in zip(weights, values):
        for w in range(capacity, weight - 1, -1):
            dp[w] = max(dp[w], value + dp[w - weight])

    return dp[capacity]


print(knapsack_01_optimized(weights, values, capacity))
print(knapsack_01_optimized(weights, values, 3))
print(knapsack_01_optimized(weights, values, 0))
````

## 96. Introduction to Divide and Conquer

Divide and Conquer is an algorithm design technique that solves a problem by dividing it into smaller subproblems.

It generally follows three steps:

1. **Divide:** Break the original problem into smaller subproblems.
2. **Conquer:** Solve the subproblems, often recursively.
3. **Combine:** Combine the subproblem results to obtain the final solution.

Merge Sort and Quick Sort are common examples of algorithms that use this technique.

Divide and Conquer is especially useful when smaller subproblems can be solved independently and their results can be combined efficiently.

## 97. Understanding the Divide and Conquer Process

We can illustrate Divide and Conquer by splitting a list into smaller parts.

For example, the list `[8, 3, 6, 2]` can be divided into `[8, 3]` and `[6, 2]`.

Each part can then be divided further until each sublist contains a single element.

A single-element list is already sorted. The smaller sorted lists can then be combined to produce the final sorted list.

This is the central idea behind Merge Sort.

````python
def split_list(items):
    midpoint = len(items) // 2

    left = items[:midpoint]
    right = items[midpoint:]

    return left, right


numbers = [8, 3, 6, 2]

left, right = split_list(numbers)

print("Original list:", numbers)
print("Left half:", left)
print("Right half:", right)
````

## 98. Implementing Merge Sort

Merge Sort is a sorting algorithm based on Divide and Conquer.

It works by:
1. Dividing the list into two halves.
2. Recursively sorting each half.
3. Merging the sorted halves into one sorted list.

Merge Sort has O(n log n) time complexity in the best, average, and worst cases.

The implementation below creates new lists while splitting and merging. Its auxiliary space complexity is O(n).

````python
def merge_sort(items):
    if len(items) <= 1:
        return items

    midpoint = len(items) // 2

    left = merge_sort(items[:midpoint])
    right = merge_sort(items[midpoint:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


numbers = [8, 3, 6, 2, 7, 1, 5, 4]

print("Original list:", numbers)
print("Sorted list:", merge_sort(numbers))
````

## 99. Understanding the Merge Operation

The merge operation combines two already-sorted lists into one sorted list.

It compares the first unprocessed element of each list and selects the smaller one.

After one list is exhausted, the remaining elements of the other list are appended.

Because each element is processed once, merging lists containing a total of `n` elements takes O(n) time.

````python
left = [2, 5, 8]
right = [1, 3, 7, 9]

merged = merge(left, right)

print("Left:", left)
print("Right:", right)
print("Merged:", merged)
````

## 101. Introduction to Quick Sort

Quick Sort is a sorting algorithm based on the Divide and Conquer technique.

It works by selecting a pivot element and dividing the remaining elements into groups:

- Elements smaller than the pivot.
- Elements equal to the pivot.
- Elements greater than the pivot.

Quick Sort recursively sorts the smaller and greater groups, then combines the results.

**Average Time Complexity:** O(n log n)

**Worst-case Time Complexity:** O(n²)

**Main Idea:** Choose a pivot, partition the elements, and recursively sort the partitions.

## 102. Understanding Partitioning

Partitioning is the process of dividing a list based on a pivot.

In this implementation, we create three lists:

- `less`: Elements smaller than the pivot.
- `equal`: Elements equal to the pivot.
- `greater`: Elements greater than the pivot.

Handling equal elements separately helps us avoid unnecessary recursive calls when the list contains duplicates.

````python
numbers = [8, 3, 1, 7, 4, 8, 2]

pivot = numbers[len(numbers) // 2]

less = [number for number in numbers if number < pivot]
equal = [number for number in numbers if number == pivot]
greater = [number for number in numbers if number > pivot]

print("Pivot:", pivot)
print("Less:", less)
print("Equal:", equal)
print("Greater:", greater)
````

## 103. Implementing Quick Sort

We can implement Quick Sort recursively.

The algorithm follows these steps:

1. If the list contains zero or one element, return it.
2. Select a pivot.
3. Divide the elements into `less`, `equal`, and `greater` lists.
4. Recursively sort `less` and `greater`.
5. Combine the sorted lists.

This implementation returns a new sorted list and does not modify the original list.

````python
def quick_sort(items):
    if len(items) <= 1:
        return items.copy()

    pivot = items[len(items) // 2]

    less = [item for item in items if item < pivot]
    equal = [item for item in items if item == pivot]
    greater = [item for item in items if item > pivot]

    return quick_sort(less) + equal + quick_sort(greater)


numbers = [8, 3, 1, 7, 4, 8, 2]

sorted_numbers = quick_sort(numbers)

print("Original:", numbers)
print("Sorted:", sorted_numbers)
````

- A list containing negative numbers.

````python
test_cases = [
    [5, 2, 9, 1, 5],
    [1, 2, 3, 4, 5],
    [4, 4, 2, 2, 1],
    [],
    [7],
    [-3, 0, 5, -1]
]

for test_case in test_cases:
    print(test_case, "->", quick_sort(test_case))
````

## 105. Merge Sort vs. Quick Sort

Both Merge Sort and Quick Sort use the Divide and Conquer technique, but they organize their work differently.

| Feature | Merge Sort | Quick Sort |
|---|---|---|
| Main operation | Merging sorted lists | Partitioning around a pivot |

## 106. Introduction to Binary Search Trees

A Binary Search Tree (BST) is a binary tree that organizes values according to a specific rule.

For each node:

- Values smaller than the node are placed in its left subtree.
- Values greater than the node are placed in its right subtree.
- Each child is the root of another subtree.

A BST allows us to search for, insert, and delete values.

The performance of these operations depends on the shape of the tree.

**Average Time Complexity:** O(log n)

**Worst-case Time Complexity:** O(n)

A poorly balanced BST can become similar to a linked list.

## 107. Understanding Tree Nodes

A tree is made of nodes connected by edges.

Each node in a Binary Search Tree stores:

- A value.
- A reference to its left child.
- A reference to its right child.

The root is the first node of the tree. A childless node is called a leaf.

````python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


root = TreeNode(10)
root.left = TreeNode(5)
root.right = TreeNode(15)

print("Root:", root.value)
print("Left child:", root.left.value)
print("Right child:", root.right.value)
````

## 108. Inserting Values into a BST

To insert a value into a Binary Search Tree:

1. Start at the root.
2. If the new value is smaller, move to the left child.
3. If the new value is greater, move to the right child.
4. Continue until an empty position is found.
5. Insert the new node there.

In this implementation, duplicate values are ignored.

````python
def insert(root, value):
    if root is None:
        return TreeNode(value)

    if value < root.value:
        root.left = insert(root.left, value)
    elif value > root.value:
        root.right = insert(root.right, value)

    return root


root = None

for value in [10, 5, 15, 3, 7, 12, 18]:
    root = insert(root, value)

print("Root:", root.value)
print("Left child:", root.left.value)
print("Right child:", root.right.value)
print("Left child of 5:", root.left.left.value)
print("Right child of 5:", root.left.right.value)
````

## 109. Searching in a BST

Searching in a BST uses the ordering of its values.

At each node:

- If the target equals the node value, the target is found.
- If the target is smaller, search the left subtree.
- If the target is greater, search the right subtree.
- If the current node is `None`, the target does not exist in the tree.

This approach avoids checking every node when the tree is reasonably balanced.

````python
def search(root, target):
    if root is None:
        return False

    if target == root.value:
        return True

    if target < root.value:
        return search(root.left, target)

    return search(root.right, target)


print(search(root, 7))
print(search(root, 12))
print(search(root, 20))
````

## 110. Tree Traversal: Inorder Traversal

Tree traversal means visiting the nodes of a tree in a particular order.

Inorder traversal follows this sequence:

1. Visit the left subtree.
2. Visit the current node.
3. Visit the right subtree.

For a Binary Search Tree, inorder traversal visits the values in ascending order.

````python
def inorder(root):
    if root is None:
        return []

    return (
        inorder(root.left)
        + [root.value]
        + inorder(root.right)
    )


print(inorder(root))
````

## 111. Preorder and Postorder Traversal

Preorder and Postorder are two additional tree traversal methods.

**Preorder:**
1. Visit the current node.
2. Visit the left subtree.
3. Visit the right subtree.

**Postorder:**
1. Visit the left subtree.
2. Visit the right subtree.
3. Visit the current node.

Different traversal orders are useful for different tree-processing tasks.

````python
def preorder(root):
    if root is None:
        return []

    return (
        [root.value]
        + preorder(root.left)
        + preorder(root.right)
    )


def postorder(root):
    if root is None:
        return []

    return (
        postorder(root.left)
        + postorder(root.right)
        + [root.value]
    )


print("Preorder:", preorder(root))
print("Postorder:", postorder(root))
````

## 113. Deleting Values from a BST

Deleting a node from a Binary Search Tree has three main cases:

1. **Leaf node:** The node has no children, so it can be removed directly.
2. **One child:** The node is replaced by its only child.
3. **Two children:** The node is replaced by its inorder successor, the smallest value in its right subtree.

After deletion, the tree must still satisfy the Binary Search Tree property.

## 114. Finding the Minimum Value

In a Binary Search Tree, the smallest value is located at the leftmost node.

To find it, start at the root and keep moving to the left child until no left child exists.

````python
def find_min(root):
    if root is None:
        return None

    current = root

    while current.left is not None:
        current = current.left

    return current.value


print(find_min(root))
````

## 115. Implementing BST Deletion

The delete operation first searches for the target value.

- If the target is smaller than the current value, search the left subtree.
- If the target is greater, search the right subtree.
- If the target is found, handle one of the three deletion cases.

When the node has two children, we use the smallest value in the right subtree as its replacement. We then delete the original occurrence of that successor.

````python
def delete(root, value):
    if root is None:
        return None

    if value < root.value:
        root.left = delete(root.left, value)

    elif value > root.value:
        root.right = delete(root.right, value)

    else:
        # Case 1: No left child
        if root.left is None:
            return root.right

        # Case 2: No right child
        if root.right is None:
            return root.left

        # Case 3: Two children
        successor_value = find_min_node(root.right).value
        root.value = successor_value
        root.right = delete(root.right, successor_value)

    return root


def find_min_node(root):
    current = root

    while current.left is not None:
        current = current.left

    return current
````

## 118. Introduction to Balanced Trees

A balanced search tree keeps its height relatively small as values are inserted and deleted.

Balanced trees help maintain efficient search, insertion, and deletion operations.

Examples include:

- AVL Trees.
- Red-Black Trees.

These structures use additional rules and operations to prevent the tree from becoming excessively unbalanced.

Their search, insertion, and deletion operations generally take O(log n) time.

## 119. Measuring Tree Height

The height of a tree can be measured by counting the number of edges on its longest path from the root to a leaf.

For this lesson, we define the height of an empty tree as `-1` and the height of a leaf as `0`.

The height helps us understand how balanced or unbalanced a tree is.

````python
def tree_height(root):
    if root is None:
        return -1

    left_height = tree_height(root.left)
    right_height = tree_height(root.right)

    return 1 + max(left_height, right_height)


print("Balanced example height:", tree_height(test_root))
print("Unbalanced example height:", tree_height(unbalanced_root))
````

The time complexity of Binary Search Tree operations depends on the tree height.

| Delete | O(log n) | O(n) |

A balanced BST has a height of approximately O(log n), while an unbalanced BST can have a height of O(n).

**Key takeaway:** The shape of a BST is critical to its performance.

## 121. Introduction to AVL Trees

An AVL Tree is a self-balancing Binary Search Tree.

Like a regular BST:
- Smaller values belong in the left subtree.
- Greater values belong in the right subtree.

An AVL Tree also maintains a balance condition.

For every node, the heights of its left and right subtrees can differ by at most one.

This condition helps keep the tree height proportional to O(log n), supporting efficient search, insertion, and deletion.

**Key idea:** An AVL Tree automatically restores balance when an insertion or deletion makes a node unbalanced.

## 122. Balance Factor

The balance factor of a node is the height of its left subtree minus the height of its right subtree.

Balance Factor = Left Subtree Height - Right Subtree Height

For an AVL Tree, each node must have a balance factor of -1, 0, or 1.

- `0`: Both subtrees have equal height.
- `1`: The left subtree is one level taller.
- `-1`: The right subtree is one level taller.

A balance factor outside this range indicates that rebalancing may be required.

````python
def balance_factor(root):
    if root is None:
        return 0

    return tree_height(root.left) - tree_height(root.right)


print("Root balance factor:", balance_factor(test_root))
print("Unbalanced tree root balance factor:", balance_factor(unbalanced_root))
````

## 123. Four AVL Rotation Cases

An AVL Tree uses rotations to restore balance.

There are four main cases:

1. **LL (Left-Left):** A node becomes unbalanced because a value was inserted into the left subtree of its left child. Fix it with a right rotation.
2. **RR (Right-Right):** A node becomes unbalanced because a value was inserted into the right subtree of its right child. Fix it with a left rotation.
3. **LR (Left-Right):** A value was inserted into the right subtree of the left child. First rotate the left child to the left, then rotate the unbalanced node to the right.
4. **RL (Right-Left):** A value was inserted into the left subtree of the right child. First rotate the right child to the right, then rotate the unbalanced node to the left.

Rotations change the tree structure without violating the Binary Search Tree ordering property.

## 124. Implementing Left and Right Rotations

A right rotation moves the left child upward and the original root downward to the right.

A left rotation moves the right child upward and the original root downward to the left.

These operations preserve the BST ordering property while changing the tree's shape.

We will use a separate `AVLNode` class that stores each node's height.

````python
class AVLNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.height = 1


def avl_height(node):
    if node is None:
        return 0

    return node.height


def update_height(node):
    node.height = 1 + max(
        avl_height(node.left),
        avl_height(node.right)
    )


def rotate_right(y):
    x = y.left
    subtree = x.right

    x.right = y
    y.left = subtree

    update_height(y)
    update_height(x)

    return x


def rotate_left(x):
    y = x.right
    subtree = y.left

    y.left = x
    x.right = subtree

    update_height(x)
    update_height(y)

    return y
````

## 125. Testing AVL Rotations

We can test rotations using small trees.

For a right rotation, the left child becomes the new root.

For a left rotation, the right child becomes the new root.

We will also verify that the inorder traversal remains sorted after each rotation.

````python
# Test a right rotation
right_root = AVLNode(30)
right_root.left = AVLNode(20)
right_root.left.left = AVLNode(10)

right_root.left.height = 2
right_root.height = 3

right_root = rotate_right(right_root)

print("Right rotation root:", right_root.value)
print("Right rotation left child:", right_root.left.value)
print("Right rotation right child:", right_root.right.value)


# Test a left rotation
left_root = AVLNode(10)
left_root.right = AVLNode(20)
left_root.right.right = AVLNode(30)

left_root.right.height = 2
left_root.height = 3

left_root = rotate_left(left_root)

print("Left rotation root:", left_root.value)
print("Left rotation left child:", left_root.left.value)
print("Left rotation right child:", left_root.right.value)
````

## 126. Implementing AVL Insertion

AVL insertion follows the usual BST insertion process, then restores balance if necessary.

The algorithm:

1. Insert the value using BST ordering.
2. Update the current node's height.
3. Calculate its balance factor.
4. Identify the imbalance case.
5. Apply the appropriate rotation.

The four cases are LL, RR, LR, and RL.

This implementation ignores duplicate values.

````python
def avl_balance_factor(node):
    if node is None:
        return 0

    return avl_height(node.left) - avl_height(node.right)


def avl_insert(root, value):
    if root is None:
        return AVLNode(value)

    if value < root.value:
        root.left = avl_insert(root.left, value)
    elif value > root.value:
        root.right = avl_insert(root.right, value)
    else:
        return root

    update_height(root)
    balance = avl_balance_factor(root)

    # LL case
    if balance > 1 and value < root.left.value:
        return rotate_right(root)

    # RR case
    if balance < -1 and value > root.right.value:
        return rotate_left(root)

    # LR case
    if balance > 1 and value > root.left.value:
        root.left = rotate_left(root.left)
        return rotate_right(root)

    # RL case
    if balance < -1 and value < root.right.value:
        root.right = rotate_right(root.right)
        return rotate_left(root)

    return root


def avl_inorder(root):
    if root is None:
        return []

    return (
        avl_inorder(root.left)
        + [root.value]
        + avl_inorder(root.right)
    )
````

## 127. Testing AVL Insertion

After AVL insertion, the tree should have a balanced structure.

## 128. Verifying AVL Balance

A valid AVL Tree must satisfy two properties:

1. It must follow the Binary Search Tree ordering property.
2. Every node must have a balance factor between -1 and 1.

We can recursively check the balance condition for every node.

````python
def is_avl_balanced(root):
    if root is None:
        return True

    balance = avl_balance_factor(root)

    if abs(balance) > 1:
        return False

    return (
        is_avl_balanced(root.left)
        and is_avl_balanced(root.right)
    )


avl_root = None

for value in [10, 20, 30, 40, 50, 25, 5, 35]:
    avl_root = avl_insert(avl_root, value)

print("Inorder:", avl_inorder(avl_root))
print("Height:", avl_height(avl_root))
print("AVL balanced:", is_avl_balanced(avl_root))
````

## 130. Introduction to Heaps

A heap is a specialized tree-based data structure that satisfies the heap property.

A binary heap is a complete binary tree, meaning that every level is full except possibly the last, and the last level is filled from left to right.

There are two common types:

- **Min Heap:** Every parent is less than or equal to its children. The minimum value is at the root.
- **Max Heap:** Every parent is greater than or equal to its children. The maximum value is at the root.

Heaps are commonly used to implement priority queues and sorting algorithms.

**Important:** A heap is not a Binary Search Tree. It does not require every value in the left subtree to be smaller than every value in the right subtree.

## 131. Representing a Heap with a List

A binary heap can be stored efficiently in a Python list without creating separate node objects.

For an element at index `i` using zero-based indexing:

- Left child: `2 * i + 1`
- Right child: `2 * i + 2`
- Parent: `(i - 1) // 2`, when `i > 0`

This representation works because a binary heap is a complete binary tree.

````python
heap = [50, 30, 40, 10, 20, 35]

index = 1

left_child_index = 2 * index + 1
right_child_index = 2 * index + 2
parent_index = (index - 1) // 2

print("Node:", heap[index])
print("Parent:", heap[parent_index])
print("Left child:", heap[left_child_index])
print("Right child:", heap[right_child_index])
````

## 132. Understanding Min Heap and Max Heap

In a Max Heap, every parent must be greater than or equal to its children.

In a Min Heap, every parent must be less than or equal to its children.

The following examples check the heap property by comparing each parent with its children.

````python
def is_max_heap(items):
    for index in range(len(items)):
        left = 2 * index + 1
        right = 2 * index + 2

        if left < len(items) and items[index] < items[left]:
            return False

        if right < len(items) and items[index] < items[right]:
            return False

    return True


def is_min_heap(items):
    for index in range(len(items)):
        left = 2 * index + 1
        right = 2 * index + 2

        if left < len(items) and items[index] > items[left]:
            return False

        if right < len(items) and items[index] > items[right]:
            return False

    return True


print(is_max_heap([50, 30, 40, 10, 20, 35]))
print(is_min_heap([10, 20, 15, 40, 30, 25]))
print(is_max_heap([10, 30, 20]))
````

## 133. Implementing Max Heapify

Heapify restores the Max Heap property in a subtree.

Starting from a node, we compare its value with its children. If a child is larger, we swap the node with the largest child and continue downward.

The procedure assumes that the left and right subtrees are already Max Heaps.

**Time Complexity:** O(log n)

````python
def max_heapify(items, size, index):
    largest = index
    left = 2 * index + 1
    right = 2 * index + 2

    if left < size and items[left] > items[largest]:
        largest = left

    if right < size and items[right] > items[largest]:
        largest = right

    if largest != index:
        items[index], items[largest] = items[largest], items[index]
        max_heapify(items, size, largest)


numbers = [10, 30, 20, 5, 15]
max_heapify(numbers, len(numbers), 0)

print(numbers)
print(is_max_heap(numbers))
````

## 134. Building a Max Heap

We can build a Max Heap from an unsorted list by applying `max_heapify` to each non-leaf node, starting from the last non-leaf node and moving toward the root.

The last non-leaf node is located at index `n // 2 - 1`.

Building a heap this way takes O(n) time, which is more efficient than inserting every element individually in O(n log n) total time.

````python
def build_max_heap(items):
    size = len(items)

    for index in range(size // 2 - 1, -1, -1):
        max_heapify(items, size, index)

    return items


numbers = [4, 10, 3, 5, 1, 8, 7]

max_heap = build_max_heap(numbers.copy())

print(max_heap)
print(is_max_heap(max_heap))
````

## 135. Heap Sort

Heap Sort uses a Max Heap to sort values in ascending order.

The algorithm follows these steps:

1. Build a Max Heap.
2. Swap the root, which contains the largest value, with the last element in the heap.
3. Reduce the active heap size by one.
4. Restore the Max Heap property.
5. Repeat until the active heap contains one element.

**Time Complexity:** O(n log n) in the best, average, and worst cases.

**Auxiliary Space:** O(1) for this iterative, in-place implementation, excluding the input list itself.

````python
def heap_sort(items):
    size = len(items)

    build_max_heap(items)

    for end in range(size - 1, 0, -1):
        items[0], items[end] = items[end], items[0]
        max_heapify(items, end, 0)

    return items


numbers = [12, 11, 13, 5, 6, 7]

sorted_numbers = heap_sort(numbers.copy())

print("Original:", numbers)
print("Sorted:", sorted_numbers)
````

## 137. Priority Queues with Python

A priority queue removes elements according to priority rather than insertion order.

Priority queues are useful in scheduling, graph algorithms, and task management.

## 140. Introduction to Tries

A Trie, also called a Prefix Tree, is a tree-based data structure used to store and search strings.

Each edge represents a character, and a path from the root represents a sequence of characters.

Tries are useful for:

- Prefix-based searches.
- Autocomplete systems.
- Spell checkers.
- Dictionary implementations.

Unlike a Binary Search Tree, a Trie organizes information by characters rather than comparing entire strings at each node.

## 141. Understanding Trie Nodes

Each Trie node stores:

- A dictionary of child nodes.
- A Boolean flag indicating whether a complete word ends at that node.

A node can represent the end of one word and still have children representing longer words.

For example, if both `car` and `care` are stored, the node representing `r` marks the end of `car` and has a child representing `e`.

````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end_of_word = False


root = TrieNode()

root.children["c"] = TrieNode()
root.children["c"].children["a"] = TrieNode()

print(root.children.keys())
print(root.children["c"].children.keys())
````

## 142. Inserting Words into a Trie

To insert a word:

1. Start at the root.
2. Process each character.
3. If the character does not have a child node, create one.
4. Move to the child node.
5. After processing the final character, mark the node as the end of a word.

This implementation treats uppercase and lowercase letters as different characters.

````python
class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        current = self.root

        for character in word:
            if character not in current.children:
                current.children[character] = TrieNode()

            current = current.children[character]

        current.is_end_of_word = True


trie = Trie()

trie.insert("car")
trie.insert("care")
trie.insert("cat")

print("Words inserted.")
````

## 143. Searching for Words

Searching for a word requires following its characters from the root.

If a character is missing, the word does not exist.

If all characters are found, we must also check `is_end_of_word`.

This final check is important because a prefix can exist without being a complete word.

````python
def search(self, word):
    current = self.root

    for character in word:
        if character not in current.children:
            return False

        current = current.children[character]

    return current.is_end_of_word


Trie.search = search

print(trie.search("car"))
print(trie.search("care"))
print(trie.search("ca"))
print(trie.search("can"))
````

## 144. Checking Prefixes

A prefix is a sequence of characters at the beginning of a word.

For example:

- `ca` is a prefix of `car`.
- `car` is a prefix of `care`.
- `care` is not a prefix of `cat`.

Prefix search only checks whether the character path exists. It does not require the final node to mark the end of a complete word.

````python
def starts_with(self, prefix):
    current = self.root

    for character in prefix:
        if character not in current.children:
            return False

        current = current.children[character]

    return True


Trie.starts_with = starts_with

print(trie.starts_with("ca"))
print(trie.starts_with("car"))
print(trie.starts_with("can"))
print(trie.starts_with(""))
````

## 145. Implementing Autocomplete

Autocomplete suggests stored words that begin with a given prefix.

The algorithm:

1. Follow the prefix through the Trie.
2. If the prefix does not exist, return an empty list.
3. Starting from the prefix node, traverse its descendants.
4. Collect every complete word.

This implementation returns all matching words in the Trie.

````python
def autocomplete(self, prefix):
    current = self.root

    for character in prefix:
        if character not in current.children:
            return []

        current = current.children[character]

    results = []

    def collect_words(node, path):
        if node.is_end_of_word:
            results.append(path)

        for character, child in node.children.items():
            collect_words(child, path + character)

    collect_words(current, prefix)

    return results


Trie.autocomplete = autocomplete

print(trie.autocomplete("ca"))
print(trie.autocomplete("car"))
print(trie.autocomplete("do"))
````

## 153. Naive Substring Search

Substring search finds a pattern inside a larger text.

The naive algorithm checks every possible starting position and compares the pattern character by character.

If the text length is `n` and the pattern length is `m`, the worst-case time complexity is O(nm).

This approach is simple and useful for understanding more advanced pattern-matching algorithms.

````python
def naive_search(text, pattern):
    if pattern == "":
        return 0

    for start in range(len(text) - len(pattern) + 1):
        if text[start:start + len(pattern)] == pattern:
            return start

    return -1


text = "data structures and algorithms"

print(naive_search(text, "structures"))
print(naive_search(text, "algorithms"))
print(naive_search(text, "python"))
````

## 155. Introduction to the KMP Algorithm

The Knuth-Morris-Pratt (KMP) algorithm is an efficient string-matching algorithm.

Unlike naive search, KMP avoids repeatedly comparing characters that have already been matched.

It preprocesses the pattern to build an LPS array.

LPS stands for Longest Proper Prefix that is also a Suffix.

A proper prefix is a prefix that is not equal to the entire string.

The LPS array helps the algorithm determine how far the pattern can shift after a mismatch.

**Time Complexity:** O(n + m), where `n` is the text length and `m` is the pattern length.

## 156. Understanding the LPS Array

The LPS array stores the length of the longest proper prefix that is also a suffix for each prefix of the pattern.

Consider the pattern `ABABAC`.

- For `A`, the LPS value is `0`.
- For `AB`, the LPS value is `0`.
- For `ABA`, the LPS value is `1`, because `A` is both a proper prefix and a suffix.
- For `ABAB`, the LPS value is `2`, because `AB` is both a proper prefix and a suffix.

The LPS array allows KMP to reuse previous matching information.

````python
pattern = "ABABAC"

for end in range(1, len(pattern) + 1):
    prefix = pattern[:end]
    longest = 0

    for length in range(1, len(prefix)):
        if prefix[:length] == prefix[-length:]:
            longest = length

    print(prefix, "->", longest)
````

## 157. Implementing the LPS Array

The LPS (Longest Proper Prefix which is also a Suffix) array stores the length of the longest proper prefix that is also a suffix for each prefix of the pattern.

We use two variables:
- `length`: the length of the current matching prefix and suffix.
- `index`: the position currently being processed.

If characters match, we increase `length`. If they do not match, we use an earlier LPS value to avoid unnecessary comparisons.

Building the LPS array takes O(m) time, where m is the pattern length.

````python
def build_lps(pattern):
    lps = [0] * len(pattern)
    length = 0
    index = 1

    while index < len(pattern):
        if pattern[index] == pattern[length]:
            length += 1
            lps[index] = length
            index += 1
        elif length > 0:
            length = lps[length - 1]
        else:
            lps[index] = 0
            index += 1

    return lps


pattern = "ABABAC"
print(build_lps(pattern))
````

## 158. Implementing the KMP Search Algorithm

KMP searches for a pattern in a text using the LPS array.

It uses two indexes:
- `text_index`: the current position in the text.
- `pattern_index`: the current position in the pattern.

When characters match, both indexes move forward.

When characters do not match:
- If `pattern_index > 0`, KMP uses the LPS array to shift the pattern without moving backward in the text.
- Otherwise, it advances to the next text character.

The function returns the starting index of the first match, or -1 if the pattern is not found.

Time complexity: O(n + m), where n is the text length and m is the pattern length.

````python
def kmp_search(text, pattern):
    if pattern == "":
        return 0

    lps = build_lps(pattern)
    text_index = 0
    pattern_index = 0

    while text_index < len(text):
        if text[text_index] == pattern[pattern_index]:
            text_index += 1
            pattern_index += 1

            if pattern_index == len(pattern):
                return text_index - pattern_index

        elif pattern_index > 0:
            pattern_index = lps[pattern_index - 1]
        else:
            text_index += 1

    return -1


text = "data structures and algorithms"

print(kmp_search(text, "structures"))
print(kmp_search(text, "algorithms"))
print(kmp_search(text, "python"))
````

## 161. Finding All Occurrences of a Pattern

Our previous search functions return only the first occurrence of a pattern.

Sometimes, we need to find every occurrence, including overlapping matches.

For example, searching for "AA" in "AAAA" should return the starting indexes [0, 1, 2].

We will implement this using the KMP algorithm and the LPS array.

Time complexity: O(n + m + k), where:
- n is the text length.
- m is the pattern length.
- k is the number of matches.

````python
def kmp_find_all(text, pattern):
    if pattern == "":
        raise ValueError("Pattern must not be empty.")
    
    lps = build_lps(pattern)
    matches = []
    text_index = 0
    pattern_index = 0

    while text_index < len(text):
        if text[text_index] == pattern[pattern_index]:
            text_index += 1
            pattern_index += 1

            if pattern_index == len(pattern):
                matches.append(text_index - pattern_index)
                pattern_index = lps[pattern_index - 1]

        elif pattern_index > 0:
            pattern_index = lps[pattern_index - 1]
        else:
            text_index += 1

    return matches


print(kmp_find_all("AAAA", "AA"))
print(kmp_find_all("ABABABAC", "ABAC"))
print(kmp_find_all("data structures and algorithms", "a"))
````

## 163. Introduction to Edit Distance

Edit Distance measures the minimum number of single-character operations required to transform one string into another.

The standard Levenshtein distance allows three operations:

1. Insertion: Add a character.
2. Deletion: Remove a character.
3. Substitution: Replace one character with another.

For example, transforming "cat" into "cut" requires one substitution: replace "a" with "u".

Therefore, the edit distance between "cat" and "cut" is 1.

Edit Distance is useful in:
- Spell checkers.
- Approximate string matching.
- Text correction.
- Comparing names and other text values.

We can solve this problem using Dynamic Programming.

## 164. Implementing Edit Distance with Dynamic Programming

We create a two-dimensional table called `dp`.

`dp[i][j]` represents the minimum edit distance between:
- The first `i` characters of the first string.
- The first `j` characters of the second string.

Base cases:
- `dp[i][0] = i`: deleting all characters requires `i` operations.
- `dp[0][j] = j`: inserting all characters requires `j` operations.

If the current characters match, we use the diagonal value without adding an operation.

Otherwise, we take the minimum cost among deletion, insertion, and substitution, then add 1.

Time complexity: O(m * n).
Space complexity: O(m * n).

````python
def edit_distance(first, second):
    rows = len(first) + 1
    columns = len(second) + 1

    dp = [
        [0] * columns
        for _ in range(rows)
    ]

    for i in range(rows):
        dp[i][0] = i

    for j in range(columns):
        dp[0][j] = j

    for i in range(1, rows):
        for j in range(1, columns):
            if first[i - 1] == second[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                deletion = dp[i - 1][j]
                insertion = dp[i][j - 1]
                substitution = dp[i - 1][j - 1]

                dp[i][j] = 1 + min(
                    deletion,
                    insertion,
                    substitution
                )

    return dp[-1][-1]


print(edit_distance("cat", "cut"))
print(edit_distance("kitten", "sitting"))
print(edit_distance("", "abc"))
print(edit_distance("book", "book"))
````

## 168. Comparing Search Algorithms

We compare Linear Search and Binary Search.

Linear Search checks elements one by one and works on both sorted and unsorted lists.

Binary Search repeatedly halves the search interval, but it requires sorted data.

For large sorted lists, Binary Search usually requires far fewer comparisons.

````python
def linear_search(numbers, target):
    for index, number in enumerate(numbers):
        if number == target:
            return index
    return -1


def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = (left + right) // 2

        if numbers[middle] == target:
            return middle
        elif numbers[middle] < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1


numbers = [3, 7, 11, 18, 24, 31, 42, 56, 68, 79]

print("Linear Search:", linear_search(numbers, 42))
print("Binary Search:", binary_search(numbers, 42))
print("Missing value:", binary_search(numbers, 50))
````

## 169. Comparing Sorting Algorithms

We compare Bubble Sort and Merge Sort.

Bubble Sort repeatedly swaps adjacent elements when they are out of order.

Merge Sort divides the list into smaller parts, sorts them recursively, and merges the sorted parts.

Both algorithms can sort the same input, but Merge Sort generally scales better for large lists.

We use separate copies of the input so that each algorithm receives the same original data.

````python
def bubble_sort(numbers):
    numbers = numbers.copy()
    size = len(numbers)

    for i in range(size):
        swapped = False

        for j in range(0, size - i - 1):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
                swapped = True

        if not swapped:
            break

    return numbers


def merge_sort(numbers):
    if len(numbers) <= 1:
        return numbers.copy()

    middle = len(numbers) // 2
    left = merge_sort(numbers[:middle])
    right = merge_sort(numbers[middle:])

    return merge(left, right)


def merge(left, right):
    result = []
    left_index = 0
    right_index = 0

    while left_index < len(left) and right_index < len(right):
        if left[left_index] <= right[right_index]:
            result.append(left[left_index])
            left_index += 1
        else:
            result.append(right[right_index])
            right_index += 1

    result.extend(left[left_index:])
    result.extend(right[right_index:])

    return result


numbers = [38, 12, 27, 5, 19, 44, 2]

print("Bubble Sort:", bubble_sort(numbers))
print("Merge Sort:", merge_sort(numbers))
print("Original list:", numbers)
````
