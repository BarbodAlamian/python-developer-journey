## 1. What Are Data Structures?

A **data structure** is a way of organizing and storing data so that we can use it efficiently.

Different data structures are designed for different purposes.

For example:

- A list stores items in a specific order.
- A set stores unique items.
- A dictionary stores key-value pairs.
- A stack follows the Last In, First Out (LIFO) principle.
- A queue follows the First In, First Out (FIFO) principle.

Choosing the right data structure can make a program easier to understand and more efficient.

## 2. What Is an Algorithm?

An **algorithm** is a sequence of steps used to solve a problem.

An algorithm receives some input, performs a series of operations, and produces an output.

For example, an algorithm for finding the largest number in a list could:

1. Assume the first number is the largest.
2. Examine each remaining number.
3. If a larger number is found, update the current largest number.
4. Return the largest number after examining all items.

Algorithms help us solve problems in a clear and systematic way.

````python
scores = [14, 18, 11, 20, 16]

largest = scores[0]

for score in scores[1:]:
    if score > largest:
        largest = score

print("Largest score:", largest)
````

## 3. Data Structures vs. Algorithms

Data structures and algorithms are closely related, but they are not the same.

- A **data structure** organizes and stores data.
- An **algorithm** defines the steps used to solve a problem.

For example, a list can store a collection of numbers, while a searching algorithm can find a specific number in that list.

The data structure determines how the data is organized, and the algorithm determines how we process it.

## 5. What Is Time Complexity?

**Time complexity** describes how the amount of work performed by an algorithm grows as the input size increases.

It does not measure the exact execution time in seconds. Instead, it helps us understand how an algorithm behaves when it processes more data.

We usually represent the input size with the letter `n`.

For example:

- If `n = 10`, the algorithm receives 10 items.
- If `n = 100`, the algorithm receives 100 items.
- If `n = 1,000`, the algorithm receives 1,000 items.

An efficient algorithm should ideally avoid doing unnecessary work as the input size grows.

Time complexity helps us compare algorithms and choose suitable solutions for larger problems.

## 6. Common Time Complexities

### 6.1. Constant Time — O(1)

An algorithm has **constant time complexity** when the amount of work does not grow with the input size.

For example, accessing an item at a known index in a Python list takes constant time.

### 6.2. Linear Time — O(n)

An algorithm has **linear time complexity** when its amount of work grows proportionally to the input size.

For example, printing every item in a list requires visiting each item once.

### 6.3. Quadratic Time — O(n²)

An algorithm has **quadratic time complexity** when its amount of work grows proportionally to the square of the input size.

A common example is a nested loop in which each iteration of the outer loop processes all items in the inner loop.

As the input grows, a quadratic algorithm can become much slower than a linear algorithm.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
````

````python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)
````

````python
numbers = [10, 20, 30, 40, 50]

for first in numbers:
    for second in numbers:
        print(first, second)
````

## 8. How to Analyze Big O

Big O notation describes how an algorithm's work grows as the input size increases.

When analyzing Big O, we focus on the dominant growth rate rather than the exact number of operations.

We use three useful rules:

1. **Ignore constant factors.** For example, `O(2n)` simplifies to `O(n)`.
2. **Keep the fastest-growing term.** For example, `O(n² + n)` simplifies to `O(n²)`.
3. **Analyze separate loops individually.** If two loops each process `n` items one after another, their combined complexity is `O(n + n)`, which simplifies to `O(n)`.

These rules help us describe an algorithm's growth in a simple and consistent way.

## 10. Linear Search

**Linear search** is an algorithm that searches for a target value by examining items one at a time.

The algorithm:

1. Starts at the first item.
2. Compares each item with the target.
3. Returns the index if the target is found.
4. Returns `-1` if the target does not exist in the list.

Linear search works with both sorted and unsorted lists.

Its worst-case time complexity is `O(n)` because it may need to examine every item.

````python
def linear_search(numbers, target):
    for index, number in enumerate(numbers):
        if number == target:
            return index

    return -1


numbers = [12, 5, 19, 3, 8]

print(linear_search(numbers, 19))
print(linear_search(numbers, 7))
````

## 12. Binary Search

**Binary search** is an algorithm that finds a target value in a sorted list by repeatedly dividing the search range in half.

Unlike linear search, binary search requires the list to be sorted.

The algorithm:

1. Set the left and right boundaries of the search range.
2. Find the middle index.
3. Compare the middle value with the target.
4. If they are equal, return the middle index.
5. If the target is smaller, continue searching the left half.
6. If the target is larger, continue searching the right half.
7. If the search range becomes empty, return `-1`.

Binary search has a worst-case time complexity of `O(log n)`.

````python
numbers = [3, 5, 8, 12, 16, 19, 24, 31, 42]
````

````python
def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = (left + right) // 2

        if numbers[middle] == target:
            return middle

        elif numbers[middle] < target:
            left = middle + 1

        else:
            right = middle - 1

    return -1


numbers = [3, 5, 8, 12, 16, 19, 24, 31, 42]

print(binary_search(numbers, 19))
print(binary_search(numbers, 7))
````

## 13. Linear Search vs. Binary Search

Linear search and binary search solve the same problem, but they use different strategies.

| Feature | Linear Search | Binary Search |
|---|---|---|
| Data must be sorted | No | Yes |
| Search strategy | Check items one by one | Repeatedly divide the range in half |
| Worst-case time complexity | `O(n)` | `O(log n)` |
| Best suited for | Small or unsorted lists | Sorted lists |

Binary search is generally much more efficient for searching large, sorted lists.

However, sorting an unsorted list takes time too. Therefore, the cost of sorting may matter if we only need to perform a single search.

## 16. Bubble Sort

**Bubble sort** is a sorting algorithm that repeatedly compares adjacent elements and swaps them if they are in the wrong order.

For ascending order:

1. Compare the first two elements.
2. Swap them if the first element is greater than the second.
3. Move to the next pair and repeat.
4. After one complete pass, the largest unsorted element reaches the end.
5. Repeat the process until the list is sorted.

Bubble sort has an average and worst-case time complexity of `O(n²)`.

It is useful for learning sorting concepts, but it is generally inefficient for large datasets.

````python
def bubble_sort(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        for j in range(n - 1 - i):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = (
                    numbers[j + 1],
                    numbers[j]
                )

    return numbers


numbers = [5, 2, 8, 1, 3]

sorted_numbers = bubble_sort(numbers)

print("Original list:", numbers)
print("Sorted list:", sorted_numbers)
````

## 18. Selection Sort

**Selection sort** repeatedly finds the smallest element in the unsorted portion of a list and moves it to the beginning of that portion.

The algorithm:

1. Start at the first position.
2. Find the smallest element in the remaining unsorted portion.
3. Swap it with the element at the current position.
4. Move to the next position.
5. Repeat until the list is sorted.

Selection sort has `O(n²)` time complexity in the average and worst cases.

Unlike bubble sort, selection sort usually performs at most one swap per outer-loop iteration.

````python
def selection_sort(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if numbers[j] < numbers[min_index]:
                min_index = j

        numbers[i], numbers[min_index] = (
            numbers[min_index],
            numbers[i]
        )

    return numbers


numbers = [5, 2, 8, 1, 3]

print(selection_sort(numbers))
````

## 19. Bubble Sort vs. Selection Sort

Both bubble sort and selection sort have `O(n²)` average and worst-case time complexity.

Their main difference is how they organize the data:

- **Bubble sort:** Repeatedly compares adjacent elements and swaps them when necessary.
- **Selection sort:** Finds the smallest remaining element and places it in the next sorted position.

Both algorithms are useful for learning, but Python's built-in sorting tools are generally a better choice in real applications.

## 20. Insertion Sort

**Insertion sort** builds a sorted portion of a list one element at a time.

It works similarly to arranging playing cards in your hand.

The algorithm:

1. Treat the first element as the initially sorted portion.
2. Take the next element as the current value.
3. Compare it with the elements in the sorted portion, moving larger elements one position to the right.
4. Insert the current value into its correct position.
5. Repeat until every element has been processed.

Insertion sort has `O(n²)` average and worst-case time complexity. Its best-case time complexity is `O(n)` when the list is already sorted.

It can be effective for small lists or lists that are already nearly sorted.

````python
def insertion_sort(numbers):
    numbers = numbers.copy()

    for i in range(1, len(numbers)):
        current = numbers[i]
        j = i - 1

        while j >= 0 and numbers[j] > current:
            numbers[j + 1] = numbers[j]
            j -= 1

        numbers[j + 1] = current

    return numbers


numbers = [5, 2, 4, 1, 3]

print("Original list:", numbers)
print("Sorted list:", insertion_sort(numbers))
````

## 21. Comparing Sorting Algorithms

| Algorithm | Main Idea | Best Case | Average Case | Worst Case |
|---|---|---|---|---|
| Bubble Sort | Swap adjacent elements | `O(n)` with early stopping | `O(n²)` | `O(n²)` |
| Selection Sort | Find the smallest remaining element | `O(n²)` | `O(n²)` | `O(n²)` |
| Insertion Sort | Insert each element into its correct position | `O(n)` | `O(n²)` | `O(n²)` |

The best-case complexity of bubble sort shown above assumes an optimized implementation that stops when no swaps occur during a pass.

All three algorithms are useful for learning, but they are generally not the preferred choice for sorting large datasets in production applications.

````python
def bubble_sort_optimized(numbers):
    numbers = numbers.copy()
    n = len(numbers)

    for i in range(n):
        swapped = False

        for j in range(n - 1 - i):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = (
                    numbers[j + 1],
                    numbers[j]
                )
                swapped = True

        if not swapped:
            break

    return numbers


print(bubble_sort_optimized([1, 2, 3, 4, 5]))
print(bubble_sort_optimized([5, 2, 4, 1, 3]))
````

## 24. Introduction to Recursion

**Recursion** is a programming technique in which a function calls itself to solve a smaller version of the same problem.

A recursive function needs two important parts:

1. **Base case:** The condition that stops the recursion.
2. **Recursive case:** The part where the function calls itself with a smaller or simpler input.

Without a correct base case, a recursive function may continue calling itself until Python raises a `RecursionError`.

Recursion is useful for problems that can naturally be divided into smaller subproblems.

````python
def countdown(number):
    if number == 0:
        print("Done!")
        return

    print(number)
    countdown(number - 1)


countdown(3)
````

## 25. Factorial Using Recursion

The factorial of a non-negative integer `n`, written as `n!`, is the product of all positive integers from `1` to `n`.

For example:

`4! = 4 × 3 × 2 × 1 = 24`

The factorial can be defined recursively:

- `0! = 1` — the base case.
- `n! = n × (n - 1)!` for `n > 0` — the recursive case.

This definition naturally translates into a recursive Python function.

````python
def factorial(number):
    if number < 0:
        raise ValueError("Factorial is not defined for negative integers.")

    if number == 0:
        return 1

    return number * factorial(number - 1)


print(factorial(4))
print(factorial(5))
print(factorial(0))
````

## 26. Recursion vs. Iteration

**Iteration** means repeating operations using loops such as `for` and `while`.

**Recursion** means solving a problem through repeated function calls, where each call works on a smaller or simpler problem.

Both approaches can solve many of the same problems.

| Feature | Iteration | Recursion |
|---|---|---|
| Main mechanism | Loops | Function calls |
| Stopping condition | Loop condition | Base case |
| Memory usage | Often lower for simple repetition | Uses additional call-stack space |
| Common risk | Infinite loops | Excessive recursion depth |

Recursion can make some algorithms easier to express, but it is not always the most efficient solution.

````python
def factorial_iterative(number):
    if number < 0:
        raise ValueError("Factorial is not defined for negative integers.")

    result = 1

    for value in range(1, number + 1):
        result *= value

    return result


def factorial_recursive(number):
    if number < 0:
        raise ValueError("Factorial is not defined for negative integers.")

    if number == 0:
        return 1

    return number * factorial_recursive(number - 1)


print(factorial_iterative(5))
print(factorial_recursive(5))
````

## 28. Stack

A **stack** is a data structure that follows the **Last In, First Out (LIFO)** principle.

This means that the last item added to the stack is the first item removed.

Imagine a stack of plates: you place a new plate on top, and you usually remove the top plate first.

The main stack operations are:

- **Push:** Add an item to the top.
- **Pop:** Remove and return the top item.
- **Peek:** View the top item without removing it.
- **is_empty:** Check whether the stack contains any items.

In Python, a list can be used to implement a simple stack with `append()` and `pop()`.

````python
stack = []

stack.append("Page 1")
stack.append("Page 2")
stack.append("Page 3")

print("Stack:", stack)

last_item = stack.pop()

print("Removed item:", last_item)
print("Stack after pop:", stack)
````

````python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Cannot pop from an empty stack.")

        return self._items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek into an empty stack.")

        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def __len__(self):
        return len(self._items)


stack = Stack()

stack.push("A")
stack.push("B")
stack.push("C")

print("Top item:", stack.peek())
print("Removed:", stack.pop())
print("Stack size:", len(stack))
print("Is empty:", stack.is_empty())
````

## 29. Queue

A **queue** is a data structure that follows the **First In, First Out (FIFO)** principle.

This means that the first item added to the queue is the first item removed.

Imagine a line of people waiting to enter a store: the person who arrives first is served first.

The main queue operations are:

- **Enqueue:** Add an item to the back of the queue.
- **Dequeue:** Remove and return the item at the front.
- **Peek:** View the front item without removing it.
- **is_empty:** Check whether the queue contains any items.

In Python, `collections.deque` is a suitable choice for implementing a queue because adding and removing items at either end are efficient operations.

````python
from collections import deque

queue = deque()

queue.append("Customer 1")
queue.append("Customer 2")
queue.append("Customer 3")

print("Queue:", queue)

served_customer = queue.popleft()

print("Served:", served_customer)
print("Queue after dequeue:", queue)
````

````python
from collections import deque


class Queue:
    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        self._items.append(item)

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Cannot dequeue from an empty queue.")

        return self._items.popleft()

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek into an empty queue.")

        return self._items[0]

    def is_empty(self):
        return len(self._items) == 0

    def __len__(self):
        return len(self._items)


queue = Queue()

queue.enqueue("Task 1")
queue.enqueue("Task 2")
queue.enqueue("Task 3")

print("Front item:", queue.peek())
print("Processed:", queue.dequeue())
print("Queue size:", len(queue))
print("Is empty:", queue.is_empty())
````

## 32. Introduction to Linked Lists

A **linked list** is a data structure made up of individual elements called **nodes**.

Each node in a singly linked list contains two things:

- **Data:** The value stored in the node.
- **Next:** A reference to the next node in the list.

The first node is called the **head**. The last node points to `None`, indicating that there are no more nodes.

Unlike a Python list, whose elements can be accessed directly by index, a linked list is typically traversed from node to node.

Linked lists are useful for learning how references connect objects and how data structures can be built from smaller components.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


first = Node(10)
second = Node(20)
third = Node(30)

first.next = second
second.next = third

print(first.data)
print(first.next.data)
print(first.next.next.data)
print(first.next.next.next)
````

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def search(self, target):
        current = self.head
        index = 0

        while current is not None:
            if current.data == target:
                return index

            current = current.next
            index += 1

        return -1


numbers = LinkedList()

numbers.append(10)
numbers.append(20)
numbers.append(30)

print(numbers.search(20))
print(numbers.search(99))
````

## 43. Introduction to Heaps

A **Heap** is a specialized tree-based data structure that helps us access the smallest or largest element efficiently.

A heap follows a property that determines the relationship between each parent node and its children.

There are two common types:

- **Min-Heap:** The smallest element is at the root.
- **Max-Heap:** The largest element is at the root.

In a Min-Heap, every parent is less than or equal to its children. In a Max-Heap, every parent is greater than or equal to its children.

A heap is not the same as a fully sorted list. It guarantees the heap property, not that every element is sorted.

## 44. Using a Min-Heap

Python provides the built-in `heapq` module for heap operations.

In a Min-Heap, the smallest element is always available at index `0`.

The `heapq` module provides useful functions:

- `heapify()`: Converts a list into a heap.
- `heappush()`: Adds an element to a heap.
- `heappop()`: Removes and returns the smallest element.

````python
import heapq

numbers = [7, 2, 9, 1, 5]

heapq.heapify(numbers)

print(numbers)
print(numbers[0])
````

## 45. Adding and Removing Heap Elements

The `heappush()` function adds an element while preserving the heap property.

The `heappop()` function removes and returns the smallest element.

These operations allow us to repeatedly retrieve the smallest element without sorting the entire collection first.

````python
import heapq

numbers = [7, 2, 9, 1, 5]
heapq.heapify(numbers)

heapq.heappush(numbers, 3)
print(numbers[0])

smallest = heapq.heappop(numbers)

print(smallest)
print(numbers[0])
````

## 46. Priority Queues

A **Priority Queue** is a data structure in which elements are processed according to their priority rather than simply their insertion order.

A heap is commonly used to implement a priority queue.

Examples include:

- Processing urgent tasks before less urgent tasks.
- Selecting the nearest candidate in certain algorithms.
- Managing events in simulations.
- Supporting shortest-path algorithms such as Dijkstra's algorithm.

In a Min-Heap, smaller priority values are processed first.

````python
import heapq

tasks = []

heapq.heappush(tasks, (3, "Write report"))
heapq.heappush(tasks, (1, "Fix critical bug"))
heapq.heappush(tasks, (2, "Reply to email"))

while tasks:
    priority, task_name = heapq.heappop(tasks)
    print(priority, task_name)
````

We can create a function to add an edge to an undirected graph.

````python
def add_edge(graph, node_a, node_b):
    if node_a not in graph:
        graph[node_a] = []

    if node_b not in graph:
        graph[node_b] = []

    graph[node_a].append(node_b)
    graph[node_b].append(node_a)


graph = {
    "A": ["B"],
    "B": ["A"]
}

add_edge(graph, "A", "C")

print(graph)
````

## 53. Breadth-First Search (BFS)

**Breadth-First Search (BFS)** explores a graph level by level, starting from a chosen vertex.

BFS uses a queue to keep track of vertices waiting to be explored.

A `visited` set prevents the algorithm from visiting the same vertex repeatedly.

For an adjacency-list graph, BFS has a time complexity of `O(V + E)`, where:
- `V` is the number of vertices.
- `E` is the number of edges.

````python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

bfs(graph, "A")
````

## 54. Depth-First Search (DFS)

**Depth-First Search (DFS)** explores one path as deeply as possible before backtracking.

DFS can be implemented using:
- Recursion.
- An explicit stack.

A `visited` set prevents repeated visits and infinite traversal in cyclic graphs.

For an adjacency-list graph, DFS has a time complexity of `O(V + E)`.

````python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

dfs(graph, "A")
````

## 55. BFS vs. DFS

| Feature | BFS | DFS |
|---|---|---|
| Exploration strategy | Level by level | Deep along a path |
| Common implementation | Queue | Stack or recursion |
| Finds shortest path in an unweighted graph | Yes | Not guaranteed |
| Time complexity | `O(V + E)` | `O(V + E)` |
| Common uses | Shortest paths, nearby nodes | Connectivity, exploration, backtracking |

**Important:** BFS finds a shortest path in terms of the number of edges in an unweighted graph. For graphs with different edge costs, algorithms such as Dijkstra's algorithm may be needed.

## 56. Practice: Check Whether a Path Exists

Write a function named `has_path(graph, start, target)`.

The function should return:
- `True` if a path exists from `start` to `target`.
- `False` otherwise.

Use BFS to solve the problem.

Example:

```python
graph = {
    "A": ["B"],
    "B": ["A", "C"],
    "C": ["B"],
    "D": []
}
```

Expected results:

```python
has_path(graph, "A", "C")  # True
has_path(graph, "A", "D")  # False
```

````python
from collections import deque

def has_path(graph, start, target):
    if start not in graph or target not in graph:
        return False

    visited = {start}
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node == target:
            return True

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return False


graph = {
    "A": ["B"],
    "B": ["A", "C"],
    "C": ["B"],
    "D": []
}

print(has_path(graph, "A", "C"))
print(has_path(graph, "A", "D"))
````
