 ## 1. Basic Functions
 A function allows us to group a specific task into a reusable block of code.
 We define a function using the `def` keyword.
 ### Syntax
 ```python
 def function_name():
     # function body
 ```
 Defining a function does not execute it.
 We need to call the function separately.

````python
def greet():
    print("Hello, Barbod!")
````

 The function above has been defined, but it has not been executed yet.
 To execute a function, we call it using its name followed by parentheses.

````python
greet()
````

 ## 2. Parameters and Arguments
 A function can receive data from outside.
 A **parameter** is a variable defined inside the function definition.
 An **argument** is the actual value passed to the function when we call it.
 For example:
 ```python
 def greet(name):
     ...
 ```
 Here, `name` is a parameter.
 When we write:
 ```python
 greet("Barbod")
 ```
 `"Barbod"` is an argument.

````python
def greet(name):
    print(f"Hello, {name}!")
````

````python
greet("Barbod")
````

````python
greet("Ali")
````

 A parameter allows the same function to work with different input values.

 ## 3. Returning Values
 So far, our functions have printed results.
 However, in real programs we often need a function to **give a value back** to the part of the program that called it.
 Python uses the `return` statement for this purpose.
 ### Example
 ```python
 def add(a, b):
     return a + b
 ```
 The function does not print the result.
 Instead, it returns the result.

````python
def add(a, b):
    return a + b
````

 We can call the function and store the returned value in a variable.

````python
result = add(10, 20)
print(result)
````

The returned value can then be used again in another calculation.

This is one of the main differences between `print()` and `return`.

- `print()` displays a value.
- `return` sends a value back to the caller.

````python
double = result * 2
print(double)
````

 ## 6. Default Parameters
 Sometimes we want a parameter to have a default value.
 A default parameter is used when the caller does not provide a value.
 ### Example
 ```python
 def greet(name="Barbod"):
     ...
 ```
 If we call:
 ```python
 greet()
 ```
 Python uses `"Barbod"` automatically.
 If we provide another value:
 ```python
 greet("Ali")
 ```
 Python uses `"Ali"` instead.

````python
def greet_with_default(name="Barbod"):
    print(f"Hello, {name}!")
````

````python
greet_with_default()
````

````python
greet_with_default("Ali")
````

 ## 7. Practice: Power Function
 We can combine parameters, default parameters, and `return`.
 The `**` operator is used for exponentiation.
 For example:
 ```python
 5 ** 2
 ```
 means:
 ```text
 5² = 25
 ```
 Our function will use `2` as the default exponent.

````python
def power(number, exponent=2):
    return number ** exponent
````

````python
power(5)
````

````python
power(5, 3)
````

 The first call uses the default value:
 ```python
 power(5) 
 ```
 which is equivalent to:
 ```python
 power(5, 2) 
 ```
 The second call provides its own exponent:
 ```python
 power(5, 3)
 ```

## 10. Scope

Scope determines where a variable can be accessed in a program.

A variable created inside a function usually has **local scope**.

A variable created outside a function usually has **global scope**.

### Local Scope

A local variable is created inside a function and can only be accessed within that function.

````python
def calculate_square(number):
    result = number ** 2
    return result

print(calculate_square(5))
````

### Global Scope

A global variable is created outside a function.
It can be accessed from inside a function.

````python
name = "Barbod"
def greet():
    print(f"Hello, {name}!")
greet()
````

````python
city = "Tehran"
def show_city():
    print(city)
show_city()
````

## 11. The `global` Keyword

By default, assigning a value to a variable inside a function creates a local variable.

The `global` keyword allows a function to modify a global variable.

````python
count = 10
def increase_count():
    global count
    count += 2
increase_count()
count
````

## 15. Arbitrary Arguments with `*args`

`*args` allows a function to receive any number of positional arguments.

Inside the function, `args` is treated as a tuple.

````python
def calculate_sum(*numbers):
    return sum(numbers)
````

````python
print(calculate_sum(1, 2, 3))
print(calculate_sum(10, 20, 30, 40))
````

## 16. Arbitrary Keyword Arguments with `**kwargs`

`**kwargs` allows a function to receive any number of keyword arguments.

Inside the function, `kwargs` is treated as a dictionary.

````python
def show_info(**info):
    print(info)
````

````python
show_info(name = "Barbod", age = 23,  city = "Tehran")
````

## 17. Combining Parameters, `*args`, and `**kwargs`

A function can combine regular parameters, `*args`, and `**kwargs`.

The order must be:

1. Regular parameters
2. `*args`
3. `**kwargs`

````python
def show_data(name, *numbers, **info):
    print(name)
    print(numbers)
    print(info)
````

````python
show_data("Barbod", 10, 20, 30, city="Tehran", age=23)
````

## 18. Lambda Functions

## 19. Functions as Arguments

In Python, functions can be passed as arguments to other functions.

This is possible because functions are objects in Python.

````python
def apply_operation(number, operation):
    return operation(number)

def double(x):
    return x * 2
````

````python
result = apply_operation(5, double)
result
````

## 20. Higher-Order Functions

A higher-order function is a function that takes another function as an argument, returns a function, or both.

````python
def apply_twice(function, value):
    return function(function(value))

def double(x):
    return x * 2
````

````python
result = apply_twice(double, 5)
result
````

## 21. Nested Functions

A nested function is a function defined inside another function.

The inner function can be used within the outer function.

````python
def greet():
    def say_hello():
            print("Hello!")
    say_hello()
greet()
````

## 22. Closures

A closure is a function that remembers and can access variables from its enclosing function even after the enclosing function has finished executing.

````python
def create_multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply
````

````python
double = create_multiplier(2)
double(5)
````

## 23. Function Annotations

````python
# using type hints
def add(a: int, b: int) -> int:
    return a + b
````

````python
result = add(10, 20)
result
````

## 26. Function Unpacking

Python allows us to unpack a sequence or dictionary when passing arguments to a function.

`*` is used to unpack positional arguments.

`**` is used to unpack keyword arguments.

````python
def calculate_total(price, quantity):
    return price * quantity
````

````python
values = (100, 3)
calculate_total(*values)
````

````python
# for dictionaries
values = {"price": 100, "quantity": 3}
calculate_total(**values)
````

## 28. `map()`

The `map()` function applies a function to every item in an iterable.

It returns a map object, which can be converted to a list.

````python
numbers = [1, 2, 3, 4, 5]
def double(number):
    return number * 2
result = map(double, numbers)
list(result)
````

Same result with lambda

````python
numbers = [1, 2, 3, 4, 5]
result = map(lambda x: x * 2, numbers)
list(result)
````

## 29. `filter()`

The `filter()` function selects items from an iterable based on a condition.

It keeps the items for which the function returns `True`.

````python
numbers = [1, 2, 3, 4, 5, 6]
def is_even(number):
    return number % 2 == 0
result = filter(is_even, numbers)
list(result)
````

Same result with lambda

````python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
````
