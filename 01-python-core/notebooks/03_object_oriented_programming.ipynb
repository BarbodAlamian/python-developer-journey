## 1. What Is Object-Oriented Programming?

Object-Oriented Programming (OOP) is a programming approach that organizes code around **objects**.

An object combines data (attributes) and behavior (methods) into a single unit.

For example, a `Car` object may have:

- **Attributes:** `brand`, `color`, and `speed`
- **Methods:** `start()`, `stop()`, and `accelerate()`

OOP helps us organize larger programs, reuse code, and represent real-world concepts in our applications.

In Python, the main building blocks of OOP are classes and objects.

## 2. Procedural vs. Object-Oriented Programming

In procedural programming, we typically organize code around functions that operate on data.

In object-oriented programming, we organize related data and behavior into objects.

Both approaches are useful. OOP becomes especially helpful when a program contains many related entities with their own data and behavior.

```python
# Procedural approach

car_brand = "Toyota"
car_speed = 0


def accelerate(speed, amount):
    return speed + amount


car_speed = accelerate(car_speed, 20)

print(car_brand)
print(car_speed)
```

```python
# Object-oriented approach

class Car:
    def __init__(self, brand):
        self.brand = brand
        self.speed = 0

    def accelerate(self, amount):
        self.speed += amount


car = Car("Toyota")
car.accelerate(20)

print(car.brand)
print(car.speed)
```

## 3. Your First Class and Object

A **class** is a blueprint for creating objects.

An **object** is an instance of a class.

For example, `Student` can be a class, while two different students can be two separate objects created from that class.

A class defines the structure and behavior that its objects can have.

```python
class Student:
    pass


student_1 = Student()
student_2 = Student()

print(type(student_1))
print(type(student_2))
```

## 4. Attributes

Attributes are variables associated with an object or a class.

**Instance attributes** store data specific to each object.

For example, each student may have a different name and age.

Instance attributes are commonly defined using `self` inside a class method.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age


student_1 = Student("Ali", 22)
student_2 = Student("Sara", 24)

print(student_1.name)
print(student_1.age)

print(student_2.name)
print(student_2.age)
```

## 5. Methods

A method is a function defined inside a class.

Methods represent behaviors associated with objects.

For example, a `Student` object might have a method called `introduce()` that displays the student's information.

Methods can access an object's attributes through `self`.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")


student = Student("Ali", 22)
student.introduce()
```

## 6. Understanding `self`

`self` refers to the current instance of a class.

It allows a method to access the attributes and other methods of the object that calls it.

For example, when `student_1.introduce()` is called, `self` refers to `student_1`.

When `student_2.introduce()` is called, `self` refers to `student_2`.

**Important:** `self` is a convention in Python, not a reserved keyword. However, we should always use `self` as the first parameter of a regular instance method.

```python
class Student:
    def introduce(self):
        print(f"My object is {self}")


student_1 = Student()
student_2 = Student()

student_1.introduce()
student_2.introduce()
```

## 7. Understanding `__init__()`

`__init__()` is a special method that Python calls automatically when a new instance is created.

It is commonly used to initialize an object's attributes.

The parameters passed when creating an object can be received by `__init__()` and stored as instance attributes.

For example, `Student("Ali", 22)` passes `"Ali"` and `22` to the initializer.

**Note:** `__init__()` initializes an object; it is not the method that actually creates the object. Object creation is handled by `__new__()`.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Ali", 22)

print(student.name)
print(student.age)
```

## 8. Instance Attributes Can Be Different

Objects created from the same class can store different attribute values.

Changing an instance attribute usually affects only that particular object.

This allows each object to maintain its own state.

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color


car_1 = Car("Toyota", "White")
car_2 = Car("BMW", "Black")

car_1.color = "Red"

print(car_1.brand, car_1.color)
print(car_2.brand, car_2.color)
```

## 9. Methods That Modify Object State

Methods can do more than display information. They can also modify an object's attributes.

The data stored in an object at a particular moment is called its **state**.

For example, a bank account object can have a `balance` attribute and a `deposit()` method that increases the balance.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount


account = BankAccount("Ali", 1000)

account.deposit(500)

print(account.owner)
print(account.balance)
```

## 11. Encapsulation

Encapsulation is an OOP principle that groups data and related behavior within a class while controlling how the object's internal state is accessed or modified.

Python does not enforce strict private instance attributes in the same way as some other languages. Instead, it uses naming conventions and mechanisms to communicate intended access.

Common naming conventions include:

- `name`: a public attribute.
- `_name`: a non-public attribute by convention.
- `__name`: an attribute subject to name mangling.

Encapsulation helps protect an object's state and makes it easier to maintain valid data.

## 12. Public and Non-Public Attributes

A public attribute is intended to be accessed directly from outside the class.

An attribute beginning with a single underscore, such as `_balance`, is considered non-public by convention. Python does not prevent external code from accessing it, but the underscore signals that it is intended for internal use.

Double underscores trigger name mangling, which helps avoid accidental name conflicts in subclasses. They do not make an attribute completely private.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def display_balance(self):
        print(self._balance)


account = BankAccount("Ali", 1000)

print(account.owner)
account.display_balance()
```

```python
print(account._balance)
```

## 13. Using Methods to Control Changes

A class can provide methods that validate data before modifying an object's attributes.

This helps maintain valid object state.

For example, a bank account should not accept a negative deposit amount.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
            return

        self._balance += amount

    def display_balance(self):
        print(f"Balance: {self._balance}")


account = BankAccount("Ali", 1000)

account.deposit(500)
account.display_balance()

account.deposit(-200)
account.display_balance()
```

## 14. Properties with `@property`

The `@property` decorator allows a method to be accessed like an attribute.

It is useful when we want attribute-like access while keeping control over how a value is retrieved or updated.

A property can also have a setter to validate new values before storing them.

```python
class Product:
    def __init__(self, price):
        self._price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative.")

        self._price = value


product = Product(100)

print(product.price)

product.price = 150
print(product.price)
```

### Practice: Validated Bank Account

Create a `BankAccount` class with:

- A public `owner` attribute.
- A non-public `_balance` attribute.
- A `balance` property that returns the current balance.
- A `deposit(amount)` method that accepts only positive amounts.
- A `withdraw(amount)` method that rejects non-positive amounts and withdrawals greater than the available balance.

Create an account with an initial balance of `1000`, deposit `500`, withdraw `200`, and print the final balance.

```python
class BankAccount:
    def __init__(self, owner, balance):
        if balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")

        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")

        if amount > self._balance:
            raise ValueError("Insufficient funds.")

        self._balance -= amount


account = BankAccount("Ali", 1000)

account.deposit(500)
account.withdraw(200)

print(account.owner)
print(account.balance)
```

## 15. What Is Inheritance?

Inheritance is an OOP principle that allows one class to inherit attributes and methods from another class.

The class being inherited from is called the **parent class** or **base class**.

The class that inherits from it is called the **child class** or **derived class**.

Inheritance helps reuse shared behavior and create relationships between related types of objects.

For example, `Dog` and `Cat` can inherit common behavior from an `Animal` class.

```python
class Animal:
    def eat(self):
        print("This animal is eating.")


class Dog(Animal):
    pass


dog = Dog()
dog.eat()
```

## 16. Inheriting Attributes and Methods

A child class can use both the inherited methods and the inherited initialization behavior of its parent class, provided they are not overridden.

The child class can also define its own attributes and methods.

This allows related classes to share common functionality while maintaining their own specific behavior.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")


class Dog(Animal):
    def bark(self):
        print(f"{self.name} says woof!")


dog = Dog("Buddy")

dog.eat()
dog.bark()
```

## 17. Method Overriding

Method overriding occurs when a child class defines a method with the same name as a method inherited from its parent class.

When that method is called on an instance of the child class, Python uses the child's implementation.

Overriding allows child classes to customize inherited behavior.

```python
class Animal:
    def speak(self):
        print("The animal makes a sound.")


class Dog(Animal):
    def speak(self):
        print("The dog barks.")


class Cat(Animal):
    def speak(self):
        print("The cat meows.")


animal = Animal()
dog = Dog()
cat = Cat()

animal.speak()
dog.speak()
cat.speak()
```

## 18. Using `super()`

The `super()` function provides access to methods defined in the next class in the method resolution order (MRO).

It is commonly used in a child class to call a parent class's initializer or extend an inherited method.

This helps avoid duplicating initialization logic.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"My name is {self.name}.")


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def introduce(self):
        super().introduce()
        print(f"My breed is {self.breed}.")


dog = Dog("Buddy", "Golden Retriever")
dog.introduce()
```

## 19. Checking Inheritance

Python provides two useful built-in functions for checking class relationships.

- `isinstance(object, class)` checks whether an object is an instance of a class or one of its subclasses.
- `issubclass(child, parent)` checks whether one class is a subclass of another class.

```python
class Animal:
    pass


class Dog(Animal):
    pass


dog = Dog()

print(isinstance(dog, Dog))
print(isinstance(dog, Animal))

print(issubclass(Dog, Animal))
print(issubclass(Animal, Dog))
```

## 20. Multiple Inheritance

Multiple inheritance allows a class to inherit from more than one parent class.

It can be useful when a class needs to combine distinct behaviors.

However, it should be used carefully because relationships between multiple parent classes can make code harder to understand.

Python uses the **Method Resolution Order (MRO)** to determine where inherited methods are found.

```python
class Flyer:
    def fly(self):
        print("Flying...")


class Swimmer:
    def swim(self):
        print("Swimming...")


class Duck(Flyer, Swimmer):
    pass


duck = Duck()

duck.fly()
duck.swim()

print(Duck.mro())
```

## 21. What Is Polymorphism?

Polymorphism means "many forms."

In OOP, polymorphism allows objects of different classes to respond to the same method call in their own ways.

For example, different animal classes can implement a `speak()` method, but each class can produce a different sound.

Polymorphism helps us write flexible code that works with different types of objects through a common interface.

```python
class Dog:
    def speak(self):
        print("Woof!")


class Cat:
    def speak(self):
        print("Meow!")


class Cow:
    def speak(self):
        print("Moo!")


animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.speak()
```

## 22. Duck Typing

Duck typing is a Python programming style that focuses on what an object can do rather than its specific class.

The idea is often summarized as:

"If it walks like a duck and quacks like a duck, treat it like a duck."

In Python, code can work with different objects as long as they provide the required behavior.

This means inheritance is not always necessary for polymorphism.

```python
class PDFDocument:
    def print_document(self):
        print("Printing a PDF document.")


class WordDocument:
    def print_document(self):
        print("Printing a Word document.")


class ImageDocument:
    def print_document(self):
        print("Printing an image.")


def print_file(document):
    document.print_document()


print_file(PDFDocument())
print_file(WordDocument())
print_file(ImageDocument())
```

## 23. Polymorphism Through Inheritance

Polymorphism can also be implemented through inheritance and method overriding.

A parent class can define a common method, while child classes provide their own implementations.

Code that works with the parent type can then use the specialized behavior of different child objects.

```python
class Notification:
    def send(self, message):
        print(f"Sending notification: {message}")


class EmailNotification(Notification):
    def send(self, message):
        print(f"Sending email: {message}")


class SMSNotification(Notification):
    def send(self, message):
        print(f"Sending SMS: {message}")


def notify_user(notification, message):
    notification.send(message)


notify_user(EmailNotification(), "Your order has shipped.")
notify_user(SMSNotification(), "Your code is 1234.")
```

## 24. Abstract Base Classes

An abstract base class defines a common interface for its subclasses.

Using the `abc` module, we can declare abstract methods that subclasses are required to implement before their objects can be instantiated.

Abstract base classes are useful when we want to enforce a common structure across related classes.

A class with unimplemented abstract methods cannot be instantiated.

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def calculate_area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14159 * self.radius ** 2


rectangle = Rectangle(5, 3)
circle = Circle(2)

print(rectangle.calculate_area())
print(circle.calculate_area())
```

## 25. Composition vs. Inheritance

Inheritance represents an **"is-a"** relationship.

For example, a `Dog` is an `Animal`.

Composition represents a **"has-a"** relationship, where an object contains or uses another object.

For example, a `Car` has an `Engine`.

Composition is often more flexible than inheritance because components can be replaced or reused without creating deep class hierarchies.

A useful design guideline is to prefer composition when inheritance does not represent a genuine "is-a" relationship.

```python
class Engine:
    def start(self):
        print("Engine started.")


class Car:
    def __init__(self, engine):
        self.engine = engine

    def start(self):
        self.engine.start()
        print("Car is ready to drive.")


engine = Engine()
car = Car(engine)

car.start()
```

## 26. Special Methods

Special methods are methods with names that begin and end with double underscores, such as `__init__()` and `__str__()`.

They are also commonly called **dunder methods**.

Python calls many of these methods automatically when we perform operations on objects.

For example:

- `__init__()` initializes a new instance.
- `__str__()` defines a human-readable string representation.
- `__repr__()` defines a representation primarily intended for developers.
- `__len__()` defines the behavior of `len()`.
- `__eq__()` defines equality comparison using `==`.

Implementing special methods allows custom objects to work naturally with Python's built-in operations.

## 27. The `__str__()` Method

The `__str__()` method defines a readable string representation of an object.

It is commonly used when an object is passed to `print()` or converted to a string with `str()`.

The method should return a string rather than print one directly.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, {self.age} years old"


student = Student("Sara", 24)

print(student)
print(str(student))
```

## 28. The `__repr__()` Method

The `__repr__()` method defines a representation primarily intended for developers.

Ideally, the representation should be unambiguous and, when practical, contain enough information to recreate an equivalent object.

Python commonly uses `__repr__()` when displaying objects in the interactive interpreter or in collections such as lists.

If `__str__()` is not defined, Python uses `__repr__()` as a fallback for string conversion.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name={self.name!r}, age={self.age!r})"


student = Student("Sara", 24)

print(student)
print(repr(student))
print([student])
```

## 29. The `__eq__()` Method

The `__eq__()` method defines how two objects are compared using `==`.

Without a custom implementation, two separate instances of an ordinary user-defined class are generally compared by identity.

By implementing `__eq__()`, we can define equality based on an object's data instead.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        if not isinstance(other, Product):
            return NotImplemented

        return self.name == other.name and self.price == other.price


product_1 = Product("Laptop", 1000)
product_2 = Product("Laptop", 1000)
product_3 = Product("Phone", 500)

print(product_1 == product_2)
print(product_1 == product_3)
```

## 30. The `__len__()` Method

The `__len__()` method defines what `len(object)` returns.

It should return a non-negative integer representing the object's length or size.

This method is useful when a custom object represents a collection or another entity with a meaningful size.

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Ali", "Sara", "Reza"])

print(len(team))
```

## 31. Class Methods and Static Methods

Python supports three commonly used method types:

- **Instance methods** receive `self` and operate on an instance.
- **Class methods** use the `@classmethod` decorator and receive `cls`, which refers to the class.
- **Static methods** use the `@staticmethod` decorator and do not automatically receive `self` or `cls`.

Class methods are often used for alternative constructors or operations involving class-level data.

Static methods are useful for related operations that do not need automatic access to the instance or class.

```python
class Employee:
    company = "GeoTech"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

    @staticmethod
    def is_valid_name(name):
        return isinstance(name, str) and len(name.strip()) > 0


employee = Employee("Ali")

print(employee.name)
print(Employee.company)

Employee.change_company("MapTech")

print(Employee.company)
print(Employee.is_valid_name("Sara"))
print(Employee.is_valid_name(""))
```

## 32. Alternative Constructors

A class method can provide an alternative way to create objects.

This is useful when input data needs to be converted into the arguments expected by `__init__()`.

For example, a `Person` object may be created from a full name string rather than passing the first and last names separately.

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @classmethod
    def from_full_name(cls, full_name):
        first_name, last_name = full_name.split()
        return cls(first_name, last_name)


person = Person.from_full_name("Barbod Alamian")

print(person.first_name)
print(person.last_name)
```

## 33. Mutable Attributes and Shared State

Mutable objects, such as lists and dictionaries, can be modified after creation.

A common mistake is to use a mutable object as a class attribute when each instance is supposed to have its own separate data.

Class attributes are shared through the class unless an instance overrides them. Therefore, a mutable class attribute can unintentionally share data between instances.

When each object needs its own list or dictionary, initialize it as an instance attribute.

```python
class Team:
    def __init__(self):
        self.members = []


team_1 = Team()
team_2 = Team()

team_1.members.append("Ali")

print(team_1.members)
print(team_2.members)
```

```python
class Team:
    members = []  # Shared by all instances


team_1 = Team()
team_2 = Team()

team_1.members.append("Ali")

print(team_1.members)
print(team_2.members)
```

## 34. Composition in More Detail

Composition allows an object to contain or use other objects as part of its implementation.

Instead of inheriting behavior from another class, a class can delegate some of its responsibilities to a separate object.

Composition is useful when objects have a "has-a" relationship.

For example, a `Computer` has a `CPU`. The computer can use the CPU object to perform a specific task.

```python
class CPU:
    def process(self):
        print("CPU is processing instructions.")


class Computer:
    def __init__(self, cpu):
        self.cpu = cpu

    def run(self):
        self.cpu.process()
        print("Computer is running.")


cpu = CPU()
computer = Computer(cpu)

computer.run()
```

## 35. Dependency Injection

Dependency injection is a technique in which an object receives the other objects it depends on instead of creating them internally.

This makes code easier to test, reuse, and modify.

For example, a `NotificationService` can receive a notification sender through its constructor. Different sender implementations can then be supplied without changing the service itself.

```python
class EmailSender:
    def send(self, message):
        print(f"Email sent: {message}")


class SMSSender:
    def send(self, message):
        print(f"SMS sent: {message}")


class NotificationService:
    def __init__(self, sender):
        self.sender = sender

    def notify(self, message):
        self.sender.send(message)


email_sender = EmailSender()
sms_sender = SMSSender()

email_service = NotificationService(email_sender)
sms_service = NotificationService(sms_sender)

email_service.notify("Your order is ready.")
sms_service.notify("Your code is 1234.")
```

## 36. Type Hints in Classes

Type hints describe the expected types of attributes, parameters, and return values.

They improve readability and help editors and static analysis tools detect some mistakes.

Python does not generally enforce type hints at runtime by itself.

Type hints are especially useful in larger projects where multiple classes interact with one another.

```python
class Product:
    def __init__(self, name: str, price: float) -> None:
        self.name: str = name
        self.price: float = price

    def apply_discount(self, percentage: float) -> float:
        discounted_price = self.price * (1 - percentage / 100)
        return discounted_price


product = Product("Laptop", 1000.0)

print(product.name)
print(product.apply_discount(10))
```

## 37. Dataclasses

A dataclass is a convenient way to define classes that primarily store data.

The `@dataclass` decorator can automatically generate methods such as `__init__()` and `__repr__()`.

By default, dataclasses also generate an equality method, `__eq__()`.

Dataclasses are useful when a class mainly represents a record of related values rather than complex behavior.

```python
from dataclasses import dataclass


@dataclass
class Product:
    name: str
    price: float


product_1 = Product("Laptop", 1000.0)
product_2 = Product("Laptop", 1000.0)

print(product_1)
print(product_1 == product_2)
```

## 38. Common OOP Mistakes

Common mistakes when working with OOP include:

- Confusing class attributes with instance attributes.
- Forgetting to use `self` when defining an instance method.
- Forgetting to call `super().__init__()` when a child class needs to initialize its parent.
- Using inheritance when composition would be more appropriate.
- Creating mutable class attributes when each instance should have its own data.
- Making classes unnecessarily complicated.
- Writing methods that do not belong to the responsibilities of the class.

Good OOP design is not about using as many classes as possible. It is about organizing responsibilities clearly and choosing an appropriate level of abstraction.

### Practice: Library Management

Create a small library system with two classes:

**`Book`**
- Stores `title`, `author`, and `is_available`.
- Starts as available.
- Provides a readable string representation using `__str__()`.

**`Library`**
- Stores a list of books.
- Provides an `add_book(book)` method.
- Provides a `find_book(title)` method that returns a matching book or `None`.
- Provides a `borrow_book(title)` method that marks an available book as unavailable.
- Prints an appropriate message if the book does not exist or is already borrowed.

Add two books, borrow one, and display its final state.

```python
class Book:
    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author
        self.is_available = True

    def __str__(self) -> str:
        status = "Available" if self.is_available else "Borrowed"
        return f"{self.title} by {self.author} - {status}"


class Library:
    def __init__(self):
        self.books: list[Book] = []

    def add_book(self, book: Book) -> None:
        self.books.append(book)

    def find_book(self, title: str) -> Book | None:
        for book in self.books:
            if book.title == title:
                return book

        return None

    def borrow_book(self, title: str) -> None:
        book = self.find_book(title)

        if book is None:
            print("Book not found.")
            return

        if not book.is_available:
            print("Book is already borrowed.")
            return

        book.is_available = False
        print(f"You borrowed: {book.title}")


library = Library()

book_1 = Book("Clean Code", "Robert C. Martin")
book_2 = Book("Python Crash Course", "Eric Matthes")

library.add_book(book_1)
library.add_book(book_2)

library.borrow_book("Clean Code")

print(book_1)
print(book_2)
```

## 39. Chapter Summary

In this chapter, I learned the core concepts of Object-Oriented Programming in Python.

Key concepts:

- **Classes and Objects:** Classes define structure and behavior; objects are instances of classes.
- **Attributes and Methods:** Attributes store data, while methods define behavior.
- **`self` and `__init__()`:** `self` refers to the current instance, and `__init__()` initializes it.
- **Class and Instance Attributes:** Class attributes are shared through the class, while instance attributes belong to individual objects.
- **Encapsulation:** Classes can control how their internal state is accessed and modified.
- **Inheritance:** Child classes can reuse and extend parent class behavior.
- **Polymorphism:** Different objects can respond to the same method call in different ways.
- **Abstract Base Classes:** These can define interfaces that subclasses must implement.
- **Composition:** Objects can use other objects to delegate responsibilities.
- **Special Methods:** Methods such as `__str__()`, `__repr__()`, and `__eq__()` integrate custom classes with Python's built-in behavior.
- **Class and Static Methods:** These support operations associated with a class or related to it without requiring a particular instance.
- **Dataclasses:** These simplify classes that primarily store data.
- **Dependency Injection:** Objects can receive their dependencies instead of creating them internally.

OOP helps organize code into clear, reusable components. However, not every problem requires classes. Good design means choosing the simplest structure that fits the problem.

### Project Description

Our system will manage products, order items, and customer orders.

- A `Product` represents an item that can be purchased.
- An `OrderItem` represents a product and its quantity.
- An `Order` contains multiple order items and calculates the total price.

Each class will have a clear responsibility.

### Project Requirements

Our system must satisfy the following requirements:

1. A product must have a name and a non-negative price.
2. An order item must contain a product and a positive quantity.
3. The subtotal of an order item must be calculated automatically.
4. An order must store multiple order items.
5. The total order price must be calculated automatically.
6. Invalid prices and quantities must raise a `ValueError`.
7. Printing an order must display its items and total price.

```python
class Product:
    def __init__(self, name: str, price: float):
        if price < 0:
            raise ValueError("Price cannot be negative.")

        self.name = name
        self.price = price

    def __str__(self) -> str:
        return f"{self.name} (${self.price:.2f})"
```

```python
keyboard = Product("Keyboard", 40.00)
print(keyboard)
```

```python
class OrderItem:
    def __init__(self, product: Product, quantity: int):
        if quantity <= 0:
            raise ValueError("Quantity must be greater than zero.")

        self.product = product
        self.quantity = quantity

    @property
    def subtotal(self) -> float:
        return self.product.price * self.quantity

    def __str__(self) -> str:
        return (
            f"{self.product.name} x {self.quantity} "
            f"= ${self.subtotal:.2f}"
        )
```

```python
class Order:
    def __init__(self, customer_name: str):
        self.customer_name = customer_name
        self._items: list[OrderItem] = []

    def add_item(
        self,
        product: Product,
        quantity: int = 1
    ) -> None:
        item = OrderItem(product, quantity)
        self._items.append(item)

    @property
    def total(self) -> float:
        return sum(item.subtotal for item in self._items)

    def __str__(self) -> str:
        lines = [f"Order for {self.customer_name}:"]

        lines.extend(str(item) for item in self._items)
        lines.append(f"Total: ${self.total:.2f}")

        return "\n".join(lines)
```

```python
keyboard = Product("Keyboard", 40.00)
mouse = Product("Mouse", 25.00)

order = Order("Alex")

order.add_item(keyboard, quantity=2)
order.add_item(mouse, quantity=1)

print(order)
```
